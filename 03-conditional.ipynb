## Expresiones booleanas

Una *expresión booleana* es aquella cuya evaluación resulta en un valor de verdad: `True` o `False`. El operador `==` compara dos operandos y devuelve `True` si son iguales y `False` en caso contrario.

```python
>>> 5 == 5
True
>>> 5 == 6
False
```

Los valores `True` y `False` pertenecen al tipo `bool` y no son cadenas de texto.

```python
>>> type(True)
<class 'bool'>
>>> type(False)
<class 'bool'>
```

El operador `==` es uno de los *operadores de comparación*. Los demás son:

```python
x != y               # x no es igual a y
x > y                # x es mayor que y
x < y                # x es menor que y
x >= y               # x es mayor o igual que y
x <= y               # x es menor o igual que y
x is y               # x es el mismo objeto que y
x is not y           # x no es el mismo objeto que y
```

Aunque estas operaciones le resultan familiares al lector, los símbolos en Python difieren de los usados en notación matemática. Un error común es utilizar un solo signo igual (`=`) en lugar de dos (`==`). Recuerde que `=` es un operador de asignación, mientras que `==` es un operador de comparación. No existen expresiones como `=<` o `=>`.

````python
print(5 == 5)
print(5 == 6)
print(type(True))
print(type(False))
x = 7
y = 10
print(x != y)
print(x < y)
print(x >= 7)
````

## Operadores lógicos

Existen tres *operadores lógicos*: `and`, `or` y `not`. Su semántica es similar a la del lenguaje natural. Por ejemplo, la expresión

```python
x > 0 and x < 10
```

es verdadera solo si `x` es mayor que 0 y menor que 10 al mismo tiempo.

La expresión `n % 2 == 0 or n % 3 == 0` es verdadera si cualquiera de las dos condiciones se cumple; es decir, si el número es divisible por 2 o por 3.

El operador `not` niega una expresión booleana. Por ejemplo, `not (x > y)` es verdadera si la condición `x > y` es falsa.

```python
>>> x = 1
>>> y = 2
>>> x > y
False
>>> not (x > y)
True
```

En términos estrictos, los operandos de los operadores lógicos deberían ser expresiones booleanas, pero Python no es muy estricto. Cualquier número distinto de cero se interpreta como `True`.

```python
>>> 17 and True
True
```

Esta flexibilidad puede ser útil en algunos contextos, pero también introduce sutilezas que pueden resultar confusas. Es recomendable evitar este tipo de uso hasta que se domine bien el comportamiento del lenguaje.

````python
x = 5
y = 2
print(x > 0 and x < 10)
print(9 % 2 == 0 or 9 % 3 == 0)
print(not (x > y))
print(17 and True)
````

## Ejecución condicional

Para construir programas útiles, casi siempre necesitamos comprobar condiciones y modificar el flujo de ejecución en consecuencia. Las *sentencias condicionales* brindan esa capacidad. La forma más simple es la sentencia `if`:

```python
if x > 0:
    print('x es positivo')
```

La expresión booleana que aparece después de la instrucción `if` se denomina *condición*. La instrucción termina con dos puntos (`:`) y las líneas posteriores se indentan.

Si la condición es verdadera, se ejecuta la instrucción indentada. Si la condición es falsa, la instrucción se omite.

Las sentencias `if` tienen la misma estructura que las definiciones de función o los bucles `for`. Consisten en una línea de cabecera que termina en `:` y un bloque indentado. Las estructuras de este tipo se llaman *sentencias compuestas* porque abarcan más de una línea.

```python
if x > y:
    print(x)
    print(y)
```

No hay un límite en la cantidad de instrucciones que pueden aparecer dentro del bloque, pero debe haber al menos una. En algunos casos, puede resultar útil tener un bloque vacío como marcador de posición para código que aún no se ha escrito. En ese caso, se puede usar la sentencia `pass`, que hace que el intérprete de Python la acepte sin realizar ninguna acción.

```python
if x < 0:
    pass   # se debe manejar el caso de valores negativos, por ahora no hacemos nada
```

Si se escribe una sentencia `if` en el intérprete de Python, el prompt cambia de `>>>` a `...` para indicar que se ha entrado en un bloque de instrucciones.

```python
>>> x = 3
>>> if x < 10:
...    print('Pequeño')
...
Pequeño
>>>
```

Cuando se utiliza el intérprete, es necesario dejar una línea en blanco al final del bloque; de lo contrario, Python generará un error.

```python
>>> x = 3
>>> if x < 10:
...    print('Pequeño')
... print('Hecho')
  File "<stdin>", line 3
    print('Hecho')
        ^
SyntaxError: invalid syntax
```

Una línea en blanco al final del bloque no es imprescindible cuando se ejecuta un script, pero puede mejorar la legibilidad del código.

````python
x = 3
if x > 0:
    print('x es positivo')

if x < 10:
    print('x es menor que 10')

if x < 0:
    pass
````

## Ejecución alternativa

Una segunda forma de sentencia `if` es la *ejecución alternativa*, en la que existen dos posibilidades y la condición determina cuál de ellas se ejecuta. La sintaxis es la siguiente:

```python
if x % 2 == 0:
    print('x es par')
else:
    print('x es impar')
```

Si el residuo de dividir `x` entre 2 es 0, entonces `x` es par y el programa muestra un mensaje correspondiente. Si la condición es falsa, se ejecuta el segundo bloque de instrucciones.

Dado que la condición debe ser verdadera o falsa, exactamente una de las alternativas se ejecutará. Las alternativas se llaman *ramas* porque representan bifurcaciones en el flujo de ejecución.

````python
x = 8
if x % 2 == 0:
    print('x es par')
else:
    print('x es impar')
````

## Condicionales encadenados

A veces existen más de dos posibilidades y se requieren más de dos ramas. Una manera de expresar esta lógica es mediante un *condicional encadenado*:

```python
if x < y:
    print('x es menor que y')
elif x > y:
    print('x es mayor que y')
else:
    print('x e y son iguales')
```

`elif` es una abreviatura de 
. Nuevamente, exactamente una rama se ejecutará.

No existe un límite en el número de instrucciones `elif`. Si existe una cláusula `else`, debe ir al final, aunque no es obligatorio que exista.

```python
if choice == 'a':
    print('Suposición incorrecta')
elif choice == 'b':
    print('Suposición correcta')
elif choice == 'c':
    print('Casi, pero no es correcto')
```

Cada condición se evalúa en orden. Si la primera es falsa, se prueba la siguiente, y así sucesivamente. Si una de ellas es verdadera, se ejecuta la rama correspondiente y el condicional termina. Incluso cuando más de una condición es verdadera, solo se ejecuta la primera rama verdadera.

## Condicionales anidados

Un condicional puede anidarse dentro de otro. Por ejemplo, el mismo caso de tres ramas puede escribirse así:

```python
if x == y:
    print('x e y son iguales')
else:
    if x < y:
        print('x es menor que y')
    else:
        print('x es mayor que y')
```

El condicional externo tiene dos ramas. La segunda rama contiene otro `if`, que a su vez tiene dos ramas. Aunque la indentación hace evidente la estructura, los condicionales anidados se vuelven difíciles de leer con rapidez. En general, es recomendable evitarlos cuando se pueda.

Los operadores lógicos suelen ofrecer una forma más clara de simplificar condicionales anidados. Por ejemplo, el siguiente bloque:

```python
if 0 < x:
    if x < 10:
        print('x es un número positivo de un solo dígito.')
```

puede reescribirse como:

```python
if 0 < x and x < 10:
    print('x es un número positivo de un solo dígito.')
```

La instrucción `print` se ejecuta solo si ambas condiciones se cumplen al mismo tiempo.

## Captura de excepciones con `try` y `except`

Anteriormente vimos un ejemplo en el que se utilizaban `input` e `int` para leer y convertir un número entero ingresado por el usuario. También observamos lo delicado que puede resultar este proceso.

```python
>>> prompt = "¿Cuál es la velocidad del ala de una golondrina?\n"
>>> velocidad = input(prompt)
¿Cuál es la velocidad del ala de una golondrina?
¿Qué quieres decir, una golondrina africana o europea?
>>> int(velocidad)
ValueError: invalid literal for int() with base 10: '¿Qué quieres decir, una golondrina africana o europea?'
```

Cuando se ejecuta este código en el intérprete de Python, aparece un nuevo prompt y el usuario puede continuar. Sin embargo, si el mismo código se coloca en un script, la ejecución del programa se detiene inmediatamente con un traceback.

A continuación se presenta un ejemplo de un programa para convertir una temperatura en Fahrenheit a Celsius:

```python
# Ejemplo de conversión de temperatura
temp_f = float(input('Ingrese la temperatura en Fahrenheit: '))
temp_c = (temp_f - 32) * 5 / 9
print('Temperatura en Celsius:', temp_c)
```

Si se proporciona una entrada inválida, el programa falla con un mensaje poco amigable:

```text
Ingrese la temperatura en Fahrenheit: fred
Traceback (most recent call last):
  File "<stdin>", line 1, in <module>
ValueError: could not convert string to float: 'fred'
```

Python incluye una estructura de ejecución condicional para manejar este tipo de errores esperados o inesperados: `try` y `except`. La finalidad de `try` y `except` es detectar una secuencia de instrucciones que puede fallar y ejecutar un bloque alternativo si ocurre un error. Este bloque `except` se ignora si no hay error.

Se puede pensar en `try`/`except` como una 
 sobre un conjunto de instrucciones.

```python
try:
    temp_f = float(input('Ingrese la temperatura en Fahrenheit: '))
    temp_c = (temp_f - 32) * 5 / 9
    print('Temperatura en Celsius:', temp_c)
except:
    print('Por favor ingrese un número válido.')
```

Python ejecuta primero el bloque `try`. Si todo funciona correctamente, se omite la cláusula `except`. Si ocurre una excepción dentro del bloque `try`, Python sale de ese bloque y ejecuta el bloque `except`.

El manejo de excepciones se conoce como *captura de errores*. En este ejemplo, la cláusula `except` imprime un mensaje de error. En general, capturar una excepción permite corregir el problema, intentar de nuevo o terminar el programa de manera controlada.

````python
try:
    temp_f = float(input('Ingrese la temperatura en Fahrenheit: '))
    temp_c = (temp_f - 32) * 5 / 9
    print('Temperatura en Celsius:', temp_c)
except:
    print('Por favor ingrese un número válido.')
````

## Evaluación de cortocircuito de expresiones lógicas

Cuando Python evalúa una expresión lógica como `x >= 2 and (x / y) > 2`, lo hace de izquierda a derecha. Debido a la definición de `and`, si `x` es menor que 2, la expresión `x >= 2` es `False` y, por lo tanto, el resultado completo será `False`, sin importar si `(x / y) > 2` resulta en `True` o `False`.

Cuando Python detecta que no aporta nada evaluar el resto de una expresión lógica, detiene la evaluación y no realiza los cálculos restantes. A este comportamiento se le llama *cortocircuito* o *short-circuiting* en inglés.

Este comportamiento conduce a una técnica útil llamada *patrón guardián* (*guardian pattern*). Considere el siguiente ejemplo en el intérprete:

```python
>>> x = 6
>>> y = 2
>>> x >= 2 and (x / y) > 2
True
>>> x = 1
>>> y = 0
>>> x >= 2 and (x / y) > 2
False
>>> x = 6
>>> y = 0
>>> x >= 2 and (x / y) > 2
Traceback (most recent call last):
  File "<stdin>", line 1, in <module>
ZeroDivisionError: division by zero
```

El tercer cálculo falló porque Python evaluó `(x / y)` cuando `y` era cero, lo que genera un error en tiempo de ejecución. Sin embargo, en los dos primeros casos no hubo error porque la primera parte de la expresión `x >= 2` ya era suficiente para determinar el resultado.

Se puede construir la expresión lógica para poner una condición de guardia justo antes de la operación potencialmente peligrosa:

```python
>>> x = 1
>>> y = 0
>>> x >= 2 and y != 0 and (x / y) > 2
False
>>> x = 6
>>> y = 0
>>> x >= 2 and y != 0 and (x / y) > 2
False
>>> x >= 2 and (x / y) > 2 and y != 0
Traceback (most recent call last):
  File "<stdin>", line 1, in <module>
ZeroDivisionError: division by zero
```

En la primera expresión lógica, `x >= 2` es `False`, así que la evaluación se detiene en el `and`. En la segunda, `x >= 2` es `True`, pero `y != 0` es `False`, por lo que nunca se llega a `(x / y)`. En la tercera, la comprobación `y != 0` está después de `(x / y)`, por lo que aparece el error. Decimos que `y != 0` actúa como un *guardián* para asegurar que solo se ejecuta `(x / y)` si `y` es distinto de cero.

````python
x = 6
y = 0
print(x >= 2 and y != 0 and (x / y) > 2)
x = 1
print(x >= 2 and y != 0 and (x / y) > 2)
````

## Depuración

La traza que Python muestra cuando ocurre un error contiene mucha información, pero puede resultar abrumadora. Las partes más útiles suelen ser:

- el tipo de error;
- el punto en el que ocurrió.

Los errores de sintaxis suelen ser fáciles de localizar, aunque hay algunas trampas. Los errores de whitespace pueden ser difíciles porque los espacios y las tabulaciones son invisibles y tendemos a ignorarlos.

```python
>>> x = 5
>>>  y = 6
  File "<stdin>", line 1
    y = 6
    ^
IndentationError: unexpected indent
```

En este ejemplo, el problema es que la segunda línea tiene un espacio de indentación inicial. Sin embargo, el mensaje apunta a `y`, lo cual puede resultar engañoso. En general, los mensajes de error indican dónde se descubrió el problema, pero el error real puede haber ocurrido antes, incluso en una línea anterior.

En otras palabras, los mensajes de error nos dicen dónde se detectó el problema, no necesariamente dónde fue causado.

## Glosario

- **expresión booleana**: expresión cuyo valor es `True` o `False`.
- **operador de comparación**: operador que compara dos valores, como `==`, `!=`, `>`, `<`, `>=`, `<=`.
- **operador lógico**: operador que combina expresiones booleanas: `and`, `or`, `not`.
- **sentencia condicional**: instrucción que controla el flujo de ejecución según cierta condición.
- **condición**: expresión booleana en una sentencia condicional que determina qué rama se ejecuta.
- **sentencia compuesta**: instrucción que consta de cabecera y cuerpo; la cabecera termina con dos puntos y el cuerpo queda indentado.
- **rama**: una de las alternativas de ejecución dentro de una condición.
- **condicional encadenado**: una sentencia condicional con múltiples ramas alternativas.
- **condicional anidado**: sentencia condicional dentro de otra sentencia condicional.
- **traceback**: lista de funciones en ejecución que se imprime cuando ocurre una excepción.
- **cortocircuito**: comportamiento por el cual Python interrumpe la evaluación de una expresión lógica cuando ya conoce el resultado final.
- **patrón guardián**: técnica de construcción de expresiones lógicas para evitar ejecuciones potencialmente peligrosas antes de comprobar una condición de seguridad.

````python
horas = 95
tarifa = 10
if horas > 40:
    print("pago: ", horas * tarifa * 1.5)
elif horas <= 40:
    print("pago: _", horas * tarifa)

print("hecho")
````

````python
horas = input("Numero de horas")
tarifa = 10

try:
    horas = int(horas)
except:
    horas = 0 

if horas > 40:
    print("pago: ", horas * tarifa * 1.5)
elif horas <= 40:
    print("pago: _", horas * tarifa)

print("hecho")
````
