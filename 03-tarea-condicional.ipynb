## Ejercicio 2: entrada válida y mensajes de error

Modifica el programa del ejercicio 1 para que no termine con un error cuando el usuario escriba texto en lugar de un número.

### Pasos

1. Coloca las conversiones `float(input(...))` dentro de un bloque `try`.
2. Usa `except ValueError` para capturar una entrada que no se pueda convertir.
3. Imprime un mensaje claro, por ejemplo: `Por favor, ingrese valores numéricos.`
4. Coloca el cálculo del pago en el bloque `else` o después de una conversión exitosa.
5. Prueba estos casos: horas `45` y tarifa `10`; horas `cuarenta`; tarifa `diez`.

### Pregunta

Explica por qué `except ValueError` es más preciso que escribir solamente `except`. No es necesario controlar todavía valores negativos ni usar módulos adicionales.

```python
try:
    horas = float(input('Ingrese horas: '))
    tarifa = float(input('Ingrese tarifa: '))
except ValueError:
    print('Por favor, ingrese valores numéricos.')
else:
    if horas > 40:
        pago = 40 * tarifa + (horas - 40) * tarifa * 1.5
    else:
        pago = horas * tarifa
    print('Pago:', pago)

# Después de probar este ejemplo, agrega tus propias pruebas.
```
