# Gráficos de sectores <a id="6"></a>

Un `gráfico de sectores` (pie chart) es un gráfico circular que muestra proporciones numéricas dividiendo un círculo (o pastel) en rebanadas proporcionales. Es muy probable que ya conozcas los gráficos de sectores, ya que se usan ampliamente en los negocios y los medios. Podemos crear gráficos de sectores en Matplotlib pasando la palabra clave `kind=pie`.

Usemos un gráfico de sectores para explorar la proporción (porcentaje) de nuevos inmigrantes agrupados por continentes para todo el período de 1980 a 2013.


Paso 1: Reunir los datos.

Usaremos el método `groupby` de *pandas* para resumir los datos de inmigración por `Continent` (continente). El proceso general de `groupby` implica los siguientes pasos:

1.  **Dividir (Split):** dividir los datos en grupos según algún criterio.
2.  **Aplicar (Apply):** aplicar una función a cada grupo de forma independiente:
    .sum()
    .count()
    .mean()
    .std()
    .aggregate()
    .apply()
    .etc..
3.  **Combinar (Combine):** combinar los resultados en una estructura de datos.


<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/labs/Module%203/images/Mod3Fig4SplitApplyCombine.png" height="400" align="center">


````python
# agrupar los países por continentes y aplicar la función sum() 
df_continents = df_can.groupby('Continent', axis=0).sum()

# nota: el resultado del método groupby es un objeto `groupby'. 
# no podemos usarlo más hasta que apliquemos una función (p. ej. .sum())
print(type(df_can.groupby('Continent', axis=0)))

df_continents.head()
````

Paso 2: Graficar los datos. Pasaremos la palabra clave `kind = 'pie'`, junto con los siguientes parámetros adicionales:

*   `autopct` - es una cadena o función que se usa para etiquetar las porciones con su valor numérico. La etiqueta se colocará dentro de la porción. Si es una cadena de formato, la etiqueta será `fmt%pct`.
*   `startangle` - gira el inicio del gráfico de sectores en un ángulo determinado (en grados) en sentido antihorario desde el eje x.
*   `shadow` - dibuja una sombra debajo del gráfico de sectores (para dar sensación de 3D).


````python
# autopct crea los %, el ángulo de inicio representa el punto de partida
df_continents['Total'].plot(kind='pie',
                            figsize=(5, 6),
                            autopct='%1.1f%%', # añadir los porcentajes
                            startangle=90,     # ángulo de inicio 90° (África)
                            shadow=True,       # añadir sombra      
                            )

plt.title('Immigration to Canada by Continent [1980 - 2013]')
plt.axis('equal') # hace que el gráfico de sectores se vea como un círculo.
plt.legend(labels=df_continents.index, loc='upper left') 


plt.show()
````

La visualización anterior no es muy clara: los números y el texto se superponen en algunos casos. Hagamos algunas modificaciones para mejorar lo visual:

*   Eliminar las etiquetas de texto del gráfico de sectores pasando `legend` y añadirlas como una leyenda separada con `plt.legend()`.
*   Empujar los porcentajes para que queden justo fuera del gráfico de sectores pasando el parámetro `pctdistance`.
*   Pasar un conjunto personalizado de colores para los continentes con el parámetro `colors`.
*   **Explotar** el gráfico de sectores para enfatizar los tres continentes con menor proporción (África, América del Norte y América Latina y el Caribe) pasando el parámetro `explode`.


````python
colors_list = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue', 'lightgreen', 'pink']
explode_list = [0.1, 0, 0, 0, 0.1, 0.1] # proporción para cada continente con la que desplazar cada porción.

df_continents['Total'].plot(kind='pie',
                            figsize=(10, 6),
                            autopct='%1.1f%%', 
                            startangle=90,    
                            shadow=True,       
                            labels=None,         # desactivar las etiquetas en el gráfico de sectores
                            pctdistance=1.12,    # la proporción entre el centro de cada porción y el inicio del texto generado por autopct 
                            #colors=colors_list,  # añadir colores personalizados
                            #explode=explode_list # 'explotar' los 3 continentes más pequeños
                            )

# aumentar el título un 12% para que coincida con pctdistance
plt.title('Immigration to Canada by Continent [1980 - 2013]', y=1.12, fontsize = 15) 

plt.axis('equal') 

# añadir la leyenda
plt.legend(labels=df_continents.index, loc='upper left', fontsize=7) 

plt.show()
````

# Diagramas de caja <a id="8"></a>

Un `diagrama de caja` (box plot) es una forma de representar estadísticamente la *distribución* de los datos a través de cinco dimensiones principales:

*   **Mínimo:** el número más pequeño del conjunto de datos, excluyendo los valores atípicos.
*   **Primer cuartil:** el número intermedio entre el `mínimo` y la `mediana`.
*   **Segundo cuartil (Mediana):** el número intermedio del conjunto de datos (ordenado).
*   **Tercer cuartil:** el número intermedio entre la `mediana` y el `máximo`.
*   **Máximo:** el número más grande del conjunto de datos, excluyendo los valores atípicos.

Para hacer un `boxplot`, podemos usar `kind=box` en el método `plot` invocado sobre una serie o un dataframe de *pandas*.

**Subgráficos**

A menudo podemos querer graficar varios gráficos dentro de la misma figura. Por ejemplo, podríamos querer hacer una comparación lado a lado del diagrama de caja con el gráfico de líneas de la inmigración de China e India.

Para visualizar varios gráficos juntos, podemos crear una **`figure`** (lienzo general) y dividirla en **`subplots`** (subgráficos), cada uno con un gráfico. Con los **subgráficos** normalmente trabajamos con la **capa de artistas** en lugar de la **capa de scripting**.

La sintaxis típica es: <br>

```python
    fig = plt.figure() # crear la figura
    ax = fig.add_subplot(nrows, ncols, plot_number) # crear los subgráficos
```

Donde

*   `nrows` y `ncols` se usan para dividir nocionalmente la figura en (`nrows` \* `ncols`) subejes,
*   `plot_number` se usa para identificar el subgráfico particular que esta función debe crear dentro de la cuadrícula nocional. `plot_number` empieza en 1, incrementa primero a lo largo de las filas y tiene un máximo de `nrows` \* `ncols`, como se muestra a continuación.

<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/labs/Module%203/images/Mod3Fig5Subplots_V2.png" width="500" align="center">


Entonces podemos especificar en qué subgráfico colocar cada gráfico pasando el parámetro `ax` en el método `plot()` de la siguiente manera:


````python
fig = plt.figure() # crear la figura

ax0 = fig.add_subplot(1, 2, 1) # añadir el subgráfico 1 (1 fila, 2 columnas, primer gráfico)
ax1 = fig.add_subplot(1, 2, 2) # añadir el subgráfico 2 (1 fila, 2 columnas, segundo gráfico). Ver el consejo abajo**

# Subgráfico 1: diagrama de caja
df_CI.plot(kind='box', color='blue', vert=False, figsize=(20, 6), ax=ax0) # añadir al subgráfico 1
ax0.set_title('Box Plots of Immigrants from China and India (1980 - 2013)')
ax0.set_xlabel('Number of Immigrants')
ax0.set_ylabel('Countries')

# Subgráfico 2: gráfico de líneas
df_CI.plot(kind='line', figsize=(20, 6), ax=ax1) # añadir al subgráfico 2
ax1.set_title ('Line Plots of Immigrants from China and India (1980 - 2013)')
ax1.set_ylabel('Number of Immigrants')
ax1.set_xlabel('Years')

plt.show()
````

**Consejo sobre la convención de los subgráficos**

En el caso de que `nrows`, `ncols` y `plot_number` sean todos menores que 10, existe una comodidad: se puede dar un número de 3 dígitos en su lugar, donde las centenas representan `nrows`, las decenas representan `ncols` y las unidades representan `plot_number`. Por ejemplo,

```python
   subplot(211) == subplot(2, 1, 1) 
```

produce un subeje en una figura que representa el gráfico superior (es decir, el primero) en una cuadrícula nocional de 2 filas por 1 columna (en realidad no existe ninguna cuadrícula, pero conceptualmente así es como se ha posicionado el subgráfico devuelto).

# Gráficos de dispersión <a id="10"></a>

Un `gráfico de dispersión` (2D) es un método útil para comparar variables entre sí. Los gráficos de `dispersión` se parecen a los `gráficos de líneas` en que ambos mapean variables independientes y dependientes en un gráfico 2D. Mientras que en un gráfico de líneas los puntos de datos están conectados por una línea, en un gráfico de dispersión no están conectados. Los datos en un gráfico de dispersión se consideran una expresión de una tendencia. Con análisis adicionales usando herramientas como la regresión, podemos calcular matemáticamente esta relación y usarla para predecir tendencias fuera del conjunto de datos.

Empecemos explorando lo siguiente:

Usando un `gráfico de dispersión`, visualicemos la tendencia de la inmigración total a Canadá (todos los países combinados) para los años 1980 - 2013.


Paso 1: obtener el conjunto de datos. Como esperamos usar la relación entre los `años` y la `población total`, convertiremos los `años` a tipo `int`.


````python
# podemos usar el método sum() para obtener la población total por año
df_tot = pd.DataFrame(df_can[years].sum(axis=0))

# cambiar los años a tipo int (útil para la regresión más adelante)
df_tot.index = map(int, df_tot.index)

# restablecer el índice para volver a ponerlo como columna en el dataframe df_tot
df_tot.reset_index(inplace = True)

# renombrar las columnas
df_tot.columns = ['year', 'total']

# ver el dataframe final
df_tot.head()
````

Paso 2: graficar los datos. En `Matplotlib` podemos crear un gráfico de `dispersión` estableciendo `kind='scatter'` como argumento de plot. También necesitaremos pasar las palabras clave `x` e `y` para especificar las columnas que van en el eje x y en el eje y.


````python
df_tot.plot(kind='scatter', x='year', y='total', figsize=(10, 6), color='darkblue')

plt.title('Total Immigration to Canada from 1980 - 2013')
plt.xlabel('Year')
plt.ylabel('Number of Immigrants')

plt.show()
````

Observa que el gráfico de dispersión no conecta los puntos de datos entre sí. Podemos observar claramente una tendencia ascendente en los datos: a medida que pasan los años, el número total de inmigrantes aumenta. Podemos analizar matemáticamente esta tendencia ascendente usando una recta de regresión (la línea de mejor ajuste).


Así que intentemos graficar una recta lineal de mejor ajuste y usarla para predecir el número de inmigrantes en 2015.

Paso 1: obtener la ecuación de la línea de mejor ajuste. Usaremos el método `polyfit()` de **Numpy** pasando lo siguiente:

*   `x`: coordenadas x de los datos.
*   `y`: coordenadas y de los datos.
*   `deg`: grado del polinomio de ajuste. 1 = lineal, 2 = cuadrático, y así sucesivamente.


````python
x = df_tot['year']      # año en el eje x
y = df_tot['total']     # total en el eje y
fit = np.polyfit(x, y, deg=1)

fit
````

La salida es un arreglo con los coeficientes del polinomio, de mayor a menor potencia. Como estamos graficando una regresión lineal `y= a * x + b`, nuestra salida tiene 2 elementos `[5.56709228e+03, -1.09261952e+07]`, con la pendiente en la posición 0 y el intercepto en la posición 1.

Paso 2: graficar la recta de regresión sobre el `gráfico de dispersión`.


````python
df_tot.plot(kind='scatter', x='year', y='total', figsize=(10, 6), color='darkblue')

plt.title('Total Immigration to Canada from 1980 - 2013')
plt.xlabel('Year')
plt.ylabel('Number of Immigrants')

# graficar la línea de mejor ajuste
plt.plot(x, fit[0] * x + fit[1], color='red') # recuerda que x son los años
plt.annotate('y={0:.0f} x + {1:.0f}'.format(fit[0], fit[1]), xy=(2000, 150000))

plt.show()

# imprimir la línea de mejor ajuste
'No. Immigrants = {0:.0f} * Year + {1:.0f}'.format(fit[0], fit[1]) 
````

Usando la ecuación de la línea de mejor ajuste, podemos estimar el número de inmigrantes en 2015:

```python
No. Immigrants = 5567 * Year - 10926195
No. Immigrants = 5567 * 2015 - 10926195
No. Immigrants = 291,310
```

Si lo comparamos con el dato real del [Informe anual 2016](http://www.cic.gc.ca/english/resources/publications/annual-report-2016/index.asp?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkDV0101ENSkillsNetwork20297740-2021-01-01) de Ciudadanía e Inmigración de Canadá (CIC), vemos que Canadá aceptó 271,845 inmigrantes en 2015. Nuestro valor estimado de 291,310 está dentro del 7% del número real, lo cual es bastante bueno considerando que nuestros datos originales provienen de las Naciones Unidas (y podrían diferir ligeramente de los datos de CIC).

Como nota al margen, podemos observar que la inmigración tuvo una caída alrededor de 1993 - 1997. Un análisis más profundo del tema reveló que en 1993 Canadá introdujo el Proyecto de Ley C-86, que introdujo revisiones al sistema de determinación de la condición de refugiado, en su mayoría restrictivas. Enmiendas posteriores a las Regulaciones de Inmigración cancelaron el patrocinio requerido para los "parientes asistidos" y redujeron los puntos que se les otorgaban, dificultando la inmigración de familiares (distintos de la familia nuclear). Estas medidas restrictivas tuvieron un impacto directo en los números de inmigración durante los siguientes años.

# Gráficos de burbujas <a id="12"></a>

Un `gráfico de burbujas` es una variación del `gráfico de dispersión` que muestra tres dimensiones de datos (x, y, z). Los puntos de datos se reemplazan por burbujas, y el tamaño de la burbuja lo determina la tercera variable `z`, también conocida como el peso. En `maplotlib` podemos pasar un arreglo o un escalar al parámetro `s` de `plot()`, que contiene el peso de cada punto.

**Empecemos analizando el efecto de la gran depresión de Argentina**.

Argentina sufrió una gran depresión de 1998 a 2002, que causó desempleo generalizado, disturbios, la caída del gobierno y un incumplimiento de pago de la deuda externa del país. En términos de ingresos, más del 50% de los argentinos eran pobres y siete de cada diez niños argentinos eran pobres en el punto más profundo de la crisis en 2002.

Analicemos el efecto de esta crisis y comparemos la inmigración de Argentina con la de su vecino Brasil. Hagámoslo con un `gráfico de burbujas` de la inmigración de Brasil y Argentina para los años 1980 - 2013. Estableceremos los pesos de las burbujas como el valor *normalizado* de la población de cada año.


**Paso 1**: obtener los datos de Brasil y Argentina. Como en el ejemplo anterior, convertiremos los `Years` (años) a tipo int y los incluiremos en el dataframe.


````python
# dataframe transpuesto
df_can_t = df_can[years].transpose()

# convertir los Years (el índice) a tipo int
df_can_t.index = map(int, df_can_t.index)

# etiquetemos el índice. Esto será automáticamente el nombre de la columna cuando restablezcamos el índice
df_can_t.index.name = 'Year'

# restablecer el índice para traer el Year como una columna
df_can_t.reset_index(inplace=True)

# ver los cambios
df_can_t.head()
````

**Paso 2**: crear los pesos normalizados.

Hay varios métodos de normalización en estadística, cada uno con su propio uso. En este caso usaremos el [escalado de características](https://en.wikipedia.org/wiki/Feature_scaling?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkDV0101ENSkillsNetwork20297740-2021-01-01) para llevar todos los valores al rango \[0, 1]. La fórmula general es:

<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/labs/Module%203/images/Mod3Fig3FeatureScaling.png" align="center">

donde $X$ es el valor original y $X'$ es el valor normalizado correspondiente. La fórmula establece el valor máximo del conjunto de datos en 1 y el valor mínimo en 0. El resto de los puntos de datos se escalan a un valor entre 0-1 en consecuencia.


````python
# normalizar los datos de Brasil
norm_brazil = (df_can_t['Brazil'] - df_can_t['Brazil'].min()) / (df_can_t['Brazil'].max() - df_can_t['Brazil'].min())

# normalizar los datos de Argentina
norm_argentina = (df_can_t['Argentina'] - df_can_t['Argentina'].min()) / (df_can_t['Argentina'].max() - df_can_t['Argentina'].min())
````

**Paso 3**: graficar los datos.

*   Para graficar dos gráficos de dispersión distintos en un mismo gráfico, podemos incluir los ejes de un gráfico dentro del otro pasándolos mediante el parámetro `ax`.
*   También pasaremos los pesos usando el parámetro `s`. Dado que los pesos normalizados están entre 0-1, no serán visibles en el gráfico. Por lo tanto:
    *   multiplicaremos los pesos por 2000 para escalarlos en el gráfico, y,
    *   sumaremos 10 para compensar el valor mínimo (que tiene peso 0 y por lo tanto se escala con $\times 2000$).


````python
# Brasil
ax0 = df_can_t.plot(kind='scatter',
                    x='Year',
                    y='Brazil',
                    figsize=(14, 8),
                    alpha=0.5,  # transparencia
                    color='green',
                    s=norm_brazil * 2000 + 10,  # pasar los pesos 
                    xlim=(1975, 2015)
                    )

# Argentina
ax1 = df_can_t.plot(kind='scatter',
                    x='Year',
                    y='Argentina',
                    alpha=0.5,
                    color="blue",
                    s=norm_argentina * 2000 + 10,
                    ax=ax0
                    )

ax0.set_ylabel('Number of Immigrants')
ax0.set_title('Immigration from Brazil and Argentina from 1980 to 2013')
ax0.legend(['Brazil', 'Argentina'], loc='upper left', fontsize='x-large')
````

El tamaño de la burbuja corresponde a la magnitud de la población que inmigró ese año, comparada con los datos de 1980 - 2013. Cuanto más grande es la burbuja, más inmigrantes hay en ese año.

En el gráfico anterior podemos ver un aumento correspondiente en la inmigración desde Argentina durante la gran depresión de 1998 - 2002. También podemos observar un pico similar alrededor de 1985 a 1993. De hecho, Argentina había sufrido una gran depresión de 1974 a 1990, justo antes del inicio de la gran depresión de 1998 - 2002.

De forma similar, Brasil sufrió el *Efecto Samba*, cuando el real brasileño (moneda) cayó casi un 35% en 1999. Existía el temor de una crisis financiera sudamericana, ya que muchos países sudamericanos dependían en gran medida de las exportaciones industriales de Brasil. El gobierno brasileño adoptó posteriormente un programa de austeridad, y la economía se recuperó lentamente a lo largo de los años, culminando en un repunte en 2010. Los datos de inmigración reflejan estos acontecimientos.
