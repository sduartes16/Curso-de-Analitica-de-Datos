## Fundamentos de *pandas*<a id="2"></a>


Lo primero que haremos será instalar **openpyxl** (antes **xlrd**), un módulo que *pandas* necesita para leer archivos de Excel.

Lo siguiente que haremos será importar dos módulos clave de análisis de datos: *pandas* y *numpy*.


````python
import numpy as np  # útil para muchas operaciones de computación científica en Python
import pandas as pd # biblioteca principal para estructuras de datos
````

Vamos a descargar e importar nuestro conjunto de datos principal de inmigración canadiense usando el método `read_excel()` de *pandas*.


````python
df_can = pd.read_excel(
    'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Canada.xlsx',
    sheet_name='Canada by Citizenship',
    skiprows=range(20),
    skipfooter=2)

print('Data read into a pandas dataframe!')
````

Veamos las primeras 5 filas del conjunto de datos con la función `head()`.


````python
df_can.head()
# consejo: puedes indicar el número de filas que quieres ver así: df_can.head(10) 
````

También podemos ver las últimas 5 filas del conjunto de datos con la función `tail()`.


````python
df_can.tail()
````

Al analizar un conjunto de datos, siempre es buena idea empezar obteniendo información básica del dataframe. Podemos hacerlo con el método `info()`.

Este método se puede usar para obtener un resumen breve del dataframe.


````python
df_can.info(verbose=False)
````

Para obtener la lista de encabezados de las columnas podemos usar la variable de instancia `columns` del dataframe.


````python
df_can.columns
````

De forma similar, para obtener la lista de índices usamos las variables de instancia `.index`.


````python
df_can.index
````

Nota: el tipo predeterminado de las variables de instancia `index` y `columns` **NO** es `list`.


````python
print(type(df_can.columns))
print(type(df_can.index))
````

Para obtener el índice y las columnas como listas, podemos usar el método `tolist()`.


````python
df_can.columns.tolist()
````

````python
df_can.index.tolist()
````

````python
print(type(df_can.columns.tolist()))
print(type(df_can.index.tolist()))
````

Para ver las dimensiones del dataframe usamos la variable de instancia `shape`.


````python
# tamaño del dataframe (filas, columnas)
df_can.shape
````

**Nota**: los principales tipos almacenados en los objetos de *pandas* son `float`, `int`, `bool`, `datetime64[ns]`, `datetime64[ns, tz]`, `timedelta[ns]`, `category` y `object` (texto). Además, estos tipos de datos tienen tamaños de elemento, por ejemplo `int64` e `int32`.


Limpiemos el conjunto de datos para eliminar algunas columnas innecesarias. Podemos usar el método `drop()` de *pandas* de la siguiente manera:


````python
# en pandas axis=0 representa las filas (predeterminado) y axis=1 representa las columnas.
df_can.drop(['AREA','REG','DEV','Type','Coverage'], axis=1, inplace=True)
df_can.head(2)
````

Renombremos las columnas para que tengan sentido. Podemos usar el método `rename()` pasando un diccionario con los nombres antiguos y nuevos, así:


````python
df_can.rename(columns={'OdName':'Country', 'AreaName':'Continent', 'RegName':'Region'}, inplace=True)
df_can.columns
````

También añadiremos una columna 'Total' que sume el total de inmigrantes por país durante todo el período 1980 - 2013, así:


````python
df_can['Total'] = df_can.sum(axis=1)
df_can['Total']
````

Podemos comprobar cuántos objetos nulos tenemos en el conjunto de datos así:


````python
df_can.isnull().sum()
````

Por último, veamos un resumen rápido de cada columna de nuestro dataframe usando el método `describe()`.


````python
df_can.describe()
````

## *pandas* Intermedio: indexación y selección (slicing)<a id="3"></a>


### Seleccionar una columna
**Hay dos formas de filtrar por el nombre de una columna:**

Método 1: rápido y sencillo, pero solo funciona si el nombre de la columna NO tiene espacios ni caracteres especiales.
```python
    df.nombre_columna          # devuelve una serie
```

Método 2: más robusto, y permite filtrar por varias columnas.

```python
    df['columna']                  # devuelve una serie
```

```python 
    df[['columna 1', 'columna 2']]  # devuelve un dataframe
```
---


Ejemplo: intentemos filtrar por la lista de países ('Country').


````python
df_can.Country  # devuelve una serie
````

Intentemos filtrar por la lista de países ('Country') y los datos de los años 1980 - 1985.


````python
df_can[['Country', 1980, 1981, 1982, 1983, 1984, 1985]] # devuelve un dataframe
# observa que 'Country' es texto y los años son enteros. 
# por consistencia, más adelante convertiremos todos los nombres de columna a texto.
````

### Seleccionar una fila

Hay principalmente 2 formas de seleccionar filas:

```python
    df.loc[label]    # filtra por las etiquetas del índice/columna
    df.iloc[index]   # filtra por las posiciones del índice/columna
```


Antes de continuar, observa que el índice predeterminado del conjunto de datos es un rango numérico de 0 a 194. Esto dificulta mucho hacer una consulta por un país específico. Por ejemplo, para buscar datos de Japón necesitamos saber el valor del índice correspondiente.

Esto se puede arreglar muy fácilmente estableciendo la columna 'Country' como índice con el método `set_index()`.


````python
df_can.set_index('Country', inplace=True)
# consejo: la operación opuesta a set es reset. Para restablecer el índice usamos df_can.reset_index()
````

````python
df_can.head(3)
````

````python
# opcional: eliminar el nombre del índice
df_can.index.name = None
````

Ejemplo: veamos el número de inmigrantes de Japón (fila 87) para los siguientes escenarios:
    1. Todos los datos de la fila (todas las columnas)
    2. Para el año 2013
    3. Para los años 1980 a 1985


````python
# 1. todos los datos de la fila (todas las columnas)
df_can.loc['Japan']
````

````python
# métodos alternativos
df_can.iloc[87]
````

````python
df_can[df_can.index == 'Japan']
````

````python
# 2. para el año 2013
df_can.loc['Japan', 2013]
````

````python
# 3. para los años 1980 a 1985
df_can.loc['Japan', [1980, 1981, 1982, 1983, 1984, 1984]]
````

### Filtrar según un criterio <a id="4"></a>
Para filtrar el dataframe según una condición, simplemente pasamos la condición como un vector booleano. 

Por ejemplo, filtremos el dataframe para mostrar los datos de los países asiáticos (AreaName = Asia).


````python
# 1. crear la serie booleana de la condición
condition = df_can['Continent'] == 'Asia'
print(condition)
````

````python
# 2. pasar esta condición al dataFrame
df_can[condition]
````

````python
# podemos pasar varios criterios en la misma línea.
# filtremos por AreaNAme = Asia y RegName = Southern Asia

df_can[(df_can['Continent']=='Asia') & (df_can['Region']=='Southern Asia')]

# nota: al usar los operadores 'and' y 'or', pandas requiere que usemos '&' y '|' en lugar de 'and' y 'or'
# no olvides encerrar las dos condiciones entre paréntesis
````

### Ordenar los valores de un Dataframe o una Serie <a id="5"></a><br>
Puedes usar la función `sort_values()` para ordenar un DataFrame o una Serie según una o más columnas. <br>Debes especificar la(s) columna(s) por las que quieres ordenar y el orden (ascendente o descendente). A continuación la sintaxis para usarla:<br><br>
```df.sort_values(col_name, axis=0, ascending=True, inplace=False, ignore_index=False)```<br><br>
col_nam - la(s) columna(s) por las que ordenar. <br>
axis - el eje a lo largo del cual ordenar. 0 para ordenar por filas (predeterminado) y 1 para ordenar por columnas.<br>
ascending - para ordenar en orden ascendente (True, predeterminado) o descendente (False).<br>
inplace - para realizar la ordenación en el lugar (True) o devolver una copia ordenada (False, predeterminado).<br>
ignore_index - para reiniciar el índice después de ordenar (True) o mantener los valores originales del índice (False, predeterminado).<br>


Ordenemos nuestro dataframe df_can por la columna 'Total', en orden descendente, para averiguar los 5 países que más contribuyeron a la inmigración a Canadá. 


````python
df_can.sort_values(by='Total', ascending=False, axis=0, inplace=True)
top_5 = df_can.head(5)
top_5
````
