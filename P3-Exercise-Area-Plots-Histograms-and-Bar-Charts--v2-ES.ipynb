# Gráficos de área<a id="6"></a>


En el módulo anterior creamos un gráfico de líneas que visualizaba los 5 países que más inmigrantes aportaron a Canadá entre 1980 y 2013. Con una pequeña modificación del código podemos visualizar este gráfico como un gráfico acumulativo, también conocido como **gráfico de líneas apiladas** o **gráfico de área**.


````python
df_can.sort_values(['Total'], ascending=False, axis=0, inplace=True)

# obtener las 5 primeras entradas
df_top5 = df_can.head()

# transponer el dataframe
df_top5 = df_top5[years].transpose()

df_top5.head()
````

Los gráficos de área se apilan por defecto. Y para producir un gráfico de área apilado, cada columna debe tener valores todos positivos o todos negativos (cualquier valor `NaN`, es decir, "not a number" o no numérico, tomará el valor 0 por defecto). Para producir un gráfico no apilado, establece el parámetro `stacked` en `False`.


````python
# cambiemos los valores del índice de df_top5 a tipo entero para graficar
df_top5.index = df_top5.index.map(int)
df_top5.plot(kind='area',
             stacked=False,
             figsize=(20, 10))  # pasar una tupla (x, y) con el tamaño

plt.title('Immigration Trend of Top 5 Countries')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

plt.show()
````

El gráfico no apilado tiene una transparencia por defecto (valor alfa) de 0.5. Podemos modificar este valor pasando el parámetro `alpha`.


````python
df_top5.plot(kind='area', 
             alpha=0.25,  # 0 - 1, valor alfa predeterminado = 0.5
             stacked=False,
             figsize=(20, 10))

plt.title('Immigration Trend of Top 5 Countries')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

plt.show()
````

### Dos tipos de graficado

Como comentamos en los videos de clase, existen dos estilos u opciones para graficar con `matplotlib`: graficar usando la capa de artistas y graficar usando la capa de scripting.

**Opción 1: Capa de scripting (método procedimental) - usando matplotlib.pyplot como 'plt'**

Puedes usar `plt`, es decir, `matplotlib.pyplot`, y añadir más elementos llamando a distintos métodos de forma procedimental; por ejemplo, `plt.title(...)` para añadir un título o `plt.xlabel(...)` para añadir una etiqueta al eje x.

**Opción 2: Capa de artistas (método orientado a objetos) - usando una instancia de `Axes` de Matplotlib (la preferida)**

Puedes usar una instancia de `Axes` de tu gráfico actual y guardarla en una variable (por ejemplo, `ax`). Puedes añadir más elementos llamando a métodos con un pequeño cambio de sintaxis (añadiendo "`set_`" a los métodos anteriores). Por ejemplo, usa `ax.set_title()` en lugar de `plt.title()` para añadir el título, o `ax.set_xlabel()` en lugar de `plt.xlabel()` para añadir la etiqueta al eje x.

Esta opción a veces es más transparente y flexible de usar en gráficos avanzados (en particular cuando hay múltiples gráficos, como verás más adelante).

En este curso nos mantendremos en la **capa de scripting**, excepto en algunas visualizaciones avanzadas donde necesitaremos usar la **capa de artistas** para manipular aspectos avanzados de los gráficos.

# Histogramas<a id="8"></a>

Un histograma es una forma de representar la distribución de *frecuencias* de un conjunto de datos numéricos. Funciona dividiendo el eje x en *bins* (intervalos), asignando cada punto de datos de nuestro conjunto a un intervalo y luego contando cuántos puntos de datos han sido asignados a cada intervalo. Así, el eje y es la frecuencia o el número de puntos de datos en cada intervalo. Ten en cuenta que podemos cambiar el tamaño de los intervalos y normalmente hay que ajustarlo para que la distribución se muestre de forma clara.


**Pregunta:** ¿Cuál es la distribución de frecuencias del número (población) de nuevos inmigrantes de los distintos países a Canadá en 2013?


Antes de continuar con la creación del histograma, primero examinemos los datos divididos en intervalos. Para ello usaremos el método `histrogram` de **Numpy** para obtener los rangos de los intervalos y los conteos de frecuencia, así:


````python
# veamos rápidamente los datos de 2013
df_can['2013'].head()
````

````python
# np.histogram devuelve 2 valores
count, bin_edges = np.histogram(df_can['2013'])

print(count) # conteo de frecuencia
print(bin_edges) # rangos de los intervalos, por defecto = 10 intervalos
````

Por defecto, el método `histrogram` divide el conjunto de datos en 10 intervalos. La figura siguiente resume los rangos de los intervalos y la distribución de frecuencias de la inmigración en 2013. Podemos ver que en 2013:

*   178 países contribuyeron entre 0 y 3412.9 inmigrantes
*   11 países contribuyeron entre 3412.9 y 6825.8 inmigrantes
*   1 país contribuyó entre 6285.8 y 10238.7 inmigrantes, y así sucesivamente.

<img src="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/labs/Module%202/images/Mod2Fig1-Histogram.jpg" align="center" width="800">


Podemos graficar fácilmente esta distribución pasando `kind=hist` a `plot()`.


````python
df_can['2013'].plot(kind='hist', figsize=(8, 5))

# añadir un título al histograma
plt.title('Histogram of Immigration from 195 Countries in 2013')
# añadir la etiqueta del eje y
plt.ylabel('Number of Countries')
# añadir la etiqueta del eje x
plt.xlabel('Number of Immigrants')

plt.show()
````

En el gráfico anterior, el eje x representa el rango de población de inmigrantes en intervalos de 3412.9. El eje y representa el número de países que contribuyeron a esa población.

Observa que las etiquetas del eje x no coinciden con el tamaño de los intervalos. Esto se puede corregir pasando la palabra clave `xticks` con la lista de los tamaños de los intervalos, así:


````python
# 'bin_edges' es una lista de intervalos de agrupación
count, bin_edges = np.histogram(df_can['2013'])

df_can['2013'].plot(kind='hist', figsize=(8, 5), xticks=bin_edges)

plt.title('Histogram of Immigration from 195 countries in 2013') # añadir un título al histograma
plt.ylabel('Number of Countries') # añadir la etiqueta del eje y
plt.xlabel('Number of Immigrants') # añadir la etiqueta del eje x

plt.show()
````

*Nota al margen:* Podríamos usar `df_can['2013'].plot.hist()` en su lugar. De hecho, a lo largo de esta lección, usar `some_data.plot(kind='type_plot', ...)` es equivalente a `some_data.plot.type_plot(...)`. Es decir, pasar el tipo de gráfico como argumento o como método se comporta igual.

Consulta la documentación de *pandas* para más información: [http://pandas.pydata.org/pandas-docs/stable/generated/pandas.Series.plot.html](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.Series.plot.html?utm_medium=Exinfluencer&utm_source=Exinfluencer&utm_content=000026UJ&utm_term=10006555&utm_id=NA-SkillsNetwork-Channel-SkillsNetworkCoursesIBMDeveloperSkillsNetworkDV0101ENSkillsNetwork20297740-2021-01-01).

¡Eso no se ve bien!

No te preocupes: te encontrarás con situaciones como esta muchas veces al crear gráficos. La solución suele estar en cómo está estructurado el conjunto de datos subyacente.

En lugar de graficar la distribución de frecuencias de la población de los 3 países, *pandas* graficó la distribución de frecuencias de la población de los `años`.

Esto se puede arreglar fácilmente transponiendo primero el conjunto de datos y luego graficando, como se muestra a continuación.


````python
# transponer el dataframe
df_t = df_can.loc[['Denmark', 'Norway', 'Sweden'], years].transpose()
df_t.head()
````

````python
# generar el histograma
df_t.plot(kind='hist', figsize=(10, 6))

plt.title('Histogram of Immigration from Denmark, Norway, and Sweden from 1980 - 2013')
plt.ylabel('Number of Years')
plt.xlabel('Number of Immigrants')

plt.show()
````

Hagamos algunas modificaciones para mejorar el impacto y la estética del gráfico anterior:

*   aumentar el tamaño de los intervalos a 15 pasando el parámetro `bins`;
*   establecer la transparencia al 60% pasando el parámetro `alpha`;
*   etiquetar el eje x pasando el parámetro `x-label`;
*   cambiar los colores de los gráficos pasando el parámetro `color`.


````python
# obtengamos los valores de las marcas del eje x
count, bin_edges = np.histogram(df_t, 15)

# histograma no apilado
df_t.plot(kind ='hist', 
          figsize=(10, 6),
          bins=15,
          alpha=0.6,
          xticks=bin_edges,
          color=['coral', 'darkslateblue', 'mediumseagreen']
         )

plt.title('Histogram of Immigration from Denmark, Norway, and Sweden from 1980 - 2013')
plt.ylabel('Number of Years')
plt.xlabel('Number of Immigrants')

plt.show()
````

Si no queremos que los gráficos se superpongan entre sí, podemos apilarlos usando el parámetro `stacked`. Ajustemos también las etiquetas mínima y máxima del eje x para eliminar el espacio extra en los bordes del gráfico. Podemos pasar una tupla (min,max) usando el parámetro `xlim`, como se muestra a continuación.


````python
count, bin_edges = np.histogram(df_t, 15)
xmin = bin_edges[0] - 10   #  el primer valor del intervalo es 31.0, añadimos un margen de 10 por estética 
xmax = bin_edges[-1] + 10  #  el último valor del intervalo es 308.0, añadimos un margen de 10 por estética

# histograma apilado
df_t.plot(kind='hist',
          figsize=(10, 6), 
          bins=15,
          xticks=bin_edges,
          color=['coral', 'darkslateblue', 'mediumseagreen'],
          stacked=True,
          xlim=(xmin, xmax)
         )

plt.title('Histogram of Immigration from Denmark, Norway, and Sweden from 1980 - 2013')
plt.ylabel('Number of Years')
plt.xlabel('Number of Immigrants') 

plt.show()
````

# Gráficos de barras (Dataframe) <a id="10"></a>

Un gráfico de barras es una forma de representar datos donde la *longitud* de las barras representa la magnitud o el tamaño de la característica/variable. Los gráficos de barras suelen representar variables numéricas y categóricas agrupadas en intervalos.

Para crear un gráfico de barras, podemos pasar uno de dos argumentos a través del parámetro `kind` en `plot()`:

*   `kind=bar` crea un gráfico de barras *vertical*
*   `kind=barh` crea un gráfico de barras *horizontal*


**Gráfico de barras vertical**

En los gráficos de barras verticales, el eje x se usa para etiquetar y la longitud de las barras en el eje y corresponde a la magnitud de la variable medida. Los gráficos de barras verticales son especialmente útiles para analizar datos de series temporales. Una desventaja es que carecen de espacio para etiquetas de texto al pie de cada barra.

**Empecemos analizando el efecto de la crisis financiera de Islandia:**

La crisis financiera islandesa de 2008 - 2011 fue un gran acontecimiento económico y político en Islandia. En relación con el tamaño de su economía, el colapso bancario sistémico de Islandia fue el mayor experimentado por cualquier país en la historia económica. La crisis provocó una grave depresión económica en 2008 - 2011 y una importante inestabilidad política.

**Pregunta:** Comparemos el número de inmigrantes islandeses (país = 'Iceland') a Canadá desde 1980 hasta 2013.


````python
# paso 1: obtener los datos
df_iceland = df_can.loc['Iceland', years]
df_iceland.head()
````

````python
# paso 2: graficar los datos
df_iceland.plot(kind='bar', figsize=(10, 6))

plt.xlabel('Year') # añadir la etiqueta del eje x al gráfico
plt.ylabel('Number of immigrants') # añadir la etiqueta del eje y al gráfico
plt.title('Icelandic immigrants to Canada from 1980 to 2013') # añadir un título al gráfico

plt.show()
````

El gráfico de barras anterior muestra el número total de inmigrantes desglosado por año. Podemos ver claramente el impacto de la crisis financiera: el número de inmigrantes a Canadá empezó a aumentar rápidamente después de 2008.

Anotemos esto en el gráfico usando el método `annotate` de la **capa de scripting** o de la **interfaz pyplot**. Pasaremos los siguientes parámetros:

*   `s`: str, el texto de la anotación.
*   `xy`: tupla que especifica el punto (x,y) a anotar (en este caso, el punto final de la flecha).
*   `xytext`: tupla que especifica el punto (x,y) donde colocar el texto (en este caso, el punto inicial de la flecha).
*   `xycoords`: el sistema de coordenadas en el que se da xy - 'data' usa el sistema de coordenadas del objeto que se está anotando (valor predeterminado).
*   `arrowprops`: toma un diccionario de propiedades para dibujar la flecha:
    *   `arrowstyle`: especifica el estilo de la flecha; `'->'` es la flecha estándar.
    *   `connectionstyle`: especifica el tipo de conexión. `arc3` es una línea recta.
    *   `color`: especifica el color de la flecha.
    *   `lw`: especifica el grosor de la línea.

Te animo a leer la documentación de Matplotlib para más detalles sobre las anotaciones:
<https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.annotate.html>.


````python
df_iceland.plot(kind='bar', figsize=(10, 6), rot=90)  # rotar las marcas del eje x (puntos etiquetados en el eje x) 90 grados

plt.xlabel('Year')
plt.ylabel('Number of Immigrants')
plt.title('Icelandic Immigrants to Canada from 1980 to 2013')

# Anotar la flecha
plt.annotate('',  # s: str. Lo dejamos vacío para no poner texto
             xy=(32, 70),  # colocar la punta de la flecha en el punto (año 2012, pob 70)
             xytext=(28, 20),  # colocar la base de la flecha en el punto (año 2008, pob 20)
             xycoords='data',  # usará el sistema de coordenadas del objeto que se está anotando
             arrowprops=dict(arrowstyle='->', connectionstyle='arc3', color='blue', lw=2)
             )

plt.show()
````

Anotemos también un texto sobre la flecha. Pasaremos los siguientes parámetros adicionales:

*   `rotation`: ángulo de rotación del texto en grados (en sentido antihorario)
*   `va`: alineación vertical del texto \[‘center’ | ‘top’ | ‘bottom’ | ‘baseline’]
*   `ha`: alineación horizontal del texto \[‘center’ | ‘right’ | ‘left’]


````python
df_iceland.plot(kind='bar', figsize=(10, 6), rot=90)

plt.xlabel('Year')
plt.ylabel('Number of Immigrants')
plt.title('Icelandic Immigrants to Canada from 1980 to 2013')

# Anotar la flecha
plt.annotate('',  # s: str. Lo dejamos vacío para no poner texto
             xy=(32, 70),  # colocar la punta de la flecha en el punto (año 2012, pob 70)
             xytext=(28, 20),  # colocar la base de la flecha en el punto (año 2008, pob 20)
             xycoords='data',  # usará el sistema de coordenadas del objeto que se está anotando
             arrowprops=dict(arrowstyle='->', connectionstyle='arc3', color='blue', lw=2)
             )

# Anotar el texto
plt.annotate('2008 - 2011 Financial Crisis',  # texto a mostrar
             xy=(28, 30),  # iniciar el texto en el punto (año 2008, pob 30)
             rotation=72.5,  # basado en prueba y error para coincidir con la flecha
             va='bottom',  # queremos que el texto esté alineado verticalmente 'abajo'
             ha='left',  # queremos que el texto esté alineado horizontalmente 'a la izquierda'
             )

plt.show()
````

**Gráfico de barras horizontal**

A veces es más práctico representar los datos horizontalmente, sobre todo si necesitas más espacio para etiquetar las barras. En los gráficos de barras horizontales, el eje y se usa para etiquetar y la longitud de las barras en el eje x corresponde a la magnitud de la variable medida. Como verás, hay más espacio en el eje y para etiquetar variables categóricas.
