## 1. Llamadas a funciones

En el contexto de la programación, una **función** es una secuencia nombrada de sentencias que realiza un cálculo. Cuando se define una función, se especifica su nombre y la secuencia de sentencias; más tarde, se puede **"llamar"** (invocar) a la función por su nombre.

Ya hemos visto un ejemplo de **llamada a función**:

```python
type(32)
```

- El **nombre** de la función es `type`.
- La expresión entre paréntesis se denomina **argumento**: es el valor o variable que pasamos a la función como entrada.
- El resultado se denomina **valor de retorno**; para `type`, es el tipo del argumento.

Es común decir que una función *"toma"* un argumento y *"devuelve"* (retorna) un resultado.

````python
# Ejemplo de llamada a función: type() devuelve el tipo del argumento
print(type(32))        # <class 'int'>
print(type('Hola'))    # <class 'str'>
print(type(3.14159))   # <class 'float'>
````

## 2. Funciones integradas (*built-in*)

Python proporciona un conjunto de funciones integradas que podemos usar **sin necesidad de definirlas**. Los creadores de Python escribieron funciones para resolver problemas comunes y las incluyeron en el lenguaje.

Las funciones `max` y `min` devuelven el valor mayor y menor de una secuencia, respectivamente:

- `max('Hello world')` nos dice cuál es el "carácter más grande" de la cadena (resulta ser la letra `'w'`).
- `min('Hello world')` muestra el carácter más pequeño (que resulta ser un espacio `' '`).

Otra función muy común es `len`, que indica cuántos elementos contiene su argumento. Si el argumento es una cadena, devuelve el número de caracteres.

> ⚠️ **Buenas prácticas:** trate los nombres de las funciones integradas como **palabras reservadas**. Evite usar `max`, `min`, `len`, `type`, etc., como nombres de variables: al hacerlo, *sobrescribiría* la función original y perdería su funcionalidad.

````python
# Funciones integradas: max, min y len
print(max('Hello world'))   # 'w'  -> el "carácter más grande" según su código Unicode
print(min('Hello world'))   # ' '  -> el espacio es el carácter más pequeño
print(len('Hello world'))   # 11   -> número de caracteres de la cadena

# Estas funciones también operan sobre otros tipos de colecciones (lo veremos en capítulos posteriores)
print(max([3, 7, 1, 9]))    # 9
print(min([3, 7, 1, 9]))    # 1
````

## 3. Funciones de conversión de tipos

Python también proporciona funciones integradas que **convierten valores de un tipo a otro**:

- **`int()`**: convierte su argumento a entero, si puede; de lo contrario, genera un error `ValueError`. Al convertir valores de punto flotante, **no redondea**: trunca la parte fraccionaria.
- **`float()`**: convierte enteros y cadenas a números de punto flotante.
- **`str()`**: convierte su argumento a cadena de texto.

Ejemplos del comportamiento de `int`:

```python
int('32')      # 32
int('Hello')   # ValueError: invalid literal for int() with base 10: 'Hello'
int(3.99999)   # 3   (trunca, NO redondea a 4)
int(-2.3)      # -2  (trunca hacia cero)
```


````python
# Conversiones a entero
print(int('32'))      # 32   -> cadena numérica a entero
print(int(3.99999))   # 3    -> TRUNCA la parte fraccionaria (no redondea)
print(int(-2.3))      # -2   -> trunca hacia cero

# Conversiones a flotante
print(float(32))        # 32.0
print(float('3.14159')) # 3.14159

# Conversiones a cadena
print(str(32))      # '32'
print(str(3.14159)) # '3.14159'

# Conversión imposible: se captura el ValueError con try/except
try:
    valor = int('Hello')
except ValueError as e:
    print(f'Error capturado: {e}')
````

## 4. Funciones matemáticas: el módulo `math`

Python dispone del módulo **`math`**, que proporciona la mayoría de las funciones matemáticas habituales. Antes de usarlo, debemos **importarlo**:

```python
import math
```

Esta sentencia crea un **objeto módulo** llamado `math`. El objeto módulo contiene las funciones y variables definidas en él. Para acceder a una función, se especifica el nombre del módulo y el nombre de la función, separados por un punto. Este formato se conoce como **notación de punto** (*dot notation*).

Ejemplos típicos:

- `math.log10(ratio)`: logaritmo en base 10 (p. ej., para calcular decibelios a partir de la relación señal/ruido).
- `math.log(x)`: logaritmo en base $e$ (logaritmo natural).
- `math.sin(radians)`: las funciones trigonométricas (`sin`, `cos`, `tan`, ...) reciben argumentos en **radianes**.

Para convertir de grados a radianes, se divide entre 360 y se multiplica por $2\pi$:

$$\text{radianes} = \frac{\text{grados}}{360} \cdot 2\pi$$

La expresión `math.pi` obtiene la variable `pi` del módulo: una aproximación de $\pi$ con una precisión de unos 15 dígitos.

````python
import math  # crea el objeto módulo 'math'

print(math)  # <module 'math' (built-in)>

# Ejemplo 1: decibelios a partir de la relación señal/ruido
signal_power = 100.0
noise_power = 10.0
ratio = signal_power / noise_power
decibels = 10 * math.log10(ratio)
print(f'Decibelios: {decibels}')          # 10.0

# Ejemplo 2: seno de un ángulo en radianes
radians = 0.7
height = math.sin(radians)
print(f'sen(0.7 rad) = {height}')

# Conversión de grados a radianes: sen(45°)
degrees = 45
radians = degrees / 360.0 * 2 * math.pi
print(f'sen(45°) = {math.sin(radians)}')  # 0.7071067811865476

# Verificación trigonométrica: sen(45°) = sqrt(2)/2
print(f'√2 / 2   = {math.sqrt(2) / 2.0}') # 0.7071067811865476

# math.pi: aproximación de pi con ~15 dígitos de precisión
print(f'pi = {math.pi}')
````

## 5. Números aleatorios: el módulo `random`

Dados los mismos datos de entrada, la mayoría de los programas generan las mismas salidas cada vez; se dice que son **deterministas**. La determinación suele ser deseable, pues esperamos que un mismo cálculo produzca el mismo resultado. Sin embargo, para algunas aplicaciones (juegos, simulaciones, muestreo estadístico) queremos que el computador sea *impredecible*.

Hacer que un programa sea verdaderamente no determinista no es fácil, pero existen formas de que al menos *parezca* no determinista. Una de ellas es usar **algoritmos** que generan números **pseudoaleatorios**. No son verdaderamente aleatorios —son producidos por un cálculo determinista—, pero a simple vista es prácticamente imposible distinguirlos de los aleatorios.

El módulo **`random`** proporciona funciones para generar números pseudoaleatorios:

- **`random.random()`**: devuelve un flotante aleatorio entre 0.0 y 1.0 (incluye 0.0, excluye 1.0).
- **`random.randint(low, high)`**: devuelve un entero entre `low` y `high` (incluye ambos extremos).
- **`random.choice(secuencia)`**: elige un elemento al azar de una secuencia.

El módulo también ofrece funciones para generar valores de distribuciones continuas: gaussiana, exponencial, gamma, entre otras.

---

**Ejercicio 1:** Ejecute la siguiente celda en su equipo y observe los números que obtiene. Ejecútela **más de una vez** y compare los resultados entre ejecuciones.

````python
import random

# Genera 10 números pseudoaleatorios en el intervalo [0.0, 1.0)
# Ejercicio 1: ejecute esta celda varias veces y observe que los valores cambian
for i in range(10):
    x = random.random()
    print(x)
````

````python
# randint(low, high): entero aleatorio entre low y high, AMBOS incluidos
print(random.randint(5, 10))
print(random.randint(5, 10))

# choice(secuencia): elige un elemento al azar
t = [1, 2, 3]
print(random.choice(t))
print(random.choice(t))

# Reproducibilidad: con una semilla fija, la secuencia es idéntica en cada ejecución
random.seed(42)
print('Con semilla 42:', [random.randint(1, 100) for _ in range(5)])
random.seed(42)
print('Repetida:      ', [random.randint(1, 100) for _ in range(5)])  # misma secuencia
````

## 6. Definición de funciones propias

Hasta ahora solo hemos usado las funciones que vienen con Python, pero también es posible **crear funciones nuevas**. Una **definición de función** especifica el nombre de una nueva función y la secuencia de sentencias que se ejecutan cuando se la llama. Una vez definida, la función puede reutilizarse una y otra vez.

```python
def print_lyrics():
    print("Soy un leñador y estoy bien.")
    print('Duermo toda la noche y trabajo todo el día.')
```

Reglas y convenciones:

- **`def`** es la palabra clave que indica que se trata de una definición de función.
- Las reglas para los nombres de funciones son las mismas que para las variables: letras, números y algunos signos de puntuación son válidos, pero el primer carácter **no puede ser un número**. No se puede usar una palabra clave como nombre, y debe evitarse que una variable y una función compartan nombre.
- Los **paréntesis vacíos** indican que la función no recibe argumentos.
- La primera línea se llama **encabezado** (*header*) y el resto es el **cuerpo** (*body*). El encabezado debe terminar con **dos puntos** (`:`) y el cuerpo debe estar **indentado** —por convención, con **cuatro espacios**—.

Definir una función crea una **variable con el mismo nombre**, cuyo valor es un **objeto función** (de tipo `function`). Una vez definida, una función puede usarse dentro de otra: por ejemplo, `repeat_lyrics` llama dos veces a `print_lyrics`.

````python
# Definición de una función sin argumentos
def print_lyrics():
    print("Soy un leñador y estoy bien.")
    print('Duermo toda la noche y trabajo todo el día.')


# Una función puede llamar a otra función
def repeat_lyrics():
    print_lyrics()
    print_lyrics()


# La definición crea un OBJETO FUNCIÓN referenciado por una variable del mismo nombre
print(print_lyrics)        # <function print_lyrics at 0x...>
print(type(print_lyrics))  # <class 'function'>

# Llamada a la función: misma sintaxis que con las funciones integradas
repeat_lyrics()
````

## 7. Definiciones, usos y flujo de ejecución

Las **definiciones de función** se ejecutan como cualquier otra sentencia, pero su efecto es *crear objetos función*. Las sentencias del cuerpo **no se ejecutan hasta que la función es llamada**, y la definición por sí sola no genera salida.

Como era de esperarse, hay que **crear una función antes de poder ejecutarla**: la definición debe ejecutarse antes de la primera llamada.

### Flujo de ejecución

El **flujo de ejecución** es el orden en que se ejecutan las sentencias:

1. La ejecución siempre comienza en la **primera sentencia** del programa y avanza de arriba hacia abajo, una sentencia a la vez.
2. Las *definiciones* de función no alteran el flujo: solo crean el objeto función.
3. Una **llamada a función** es como un desvío: el flujo salta al cuerpo de la función, ejecuta todas sus sentencias y **regresa al punto donde quedó**.
4. Como una función puede llamar a otra (y esta a otra más), Python lleva registro de dónde está; cada vez que una función termina, el programa retoma la ejecución en la función que la llamó.

## 8. Parámetros y argumentos

Algunas funciones integradas requieren argumentos: `math.sin` recibe un número; `math.pow` recibe dos (la base y el exponente).

Dentro de la función, los argumentos se asignan a variables llamadas **parámetros**:

```python
def print_twice(bruce):
    print(bruce)
    print(bruce)
```

Esta función asigna el argumento al parámetro `bruce` e imprime su valor dos veces. Funciona con **cualquier valor que se pueda imprimir**.

### Reglas importantes

1. **Composición:** se puede usar *cualquier expresión* como argumento (`'Spam ' * 4`, `math.cos(math.pi)`). El argumento se **evalúa una sola vez**, antes de la llamada.
2. **Variables como argumentos:** se puede pasar una variable; su nombre es irrelevante dentro de la función.
3. **Independencia de nombres:** el nombre de la variable en el llamador (`michael`) no tiene nada que ver con el nombre del parámetro (`bruce`). Dentro de `print_twice`, todo el mundo se llama `bruce`.

````python
import math

# Función con UN parámetro
def print_twice(bruce):
    print(bruce)
    print(bruce)

# Funciona con cualquier valor imprimible
print_twice('Spam')       # cadena
print_twice(17)           # entero
print_twice(math.pi)      # flotante

# Composición: el argumento puede ser cualquier EXPRESIÓN (se evalúa una sola vez)
print_twice('Spam ' * 4)
print_twice(math.cos(math.pi))

# El nombre del argumento (michael) es independiente del parámetro (bruce)
michael = 'Eric, the half a bee.'
print_twice(michael)
````

## 9. Funciones fructíferas y funciones vacías

- Las **funciones fructíferas** (*fruitful*) **devuelven un resultado**: `math.sqrt`, `math.cos`, `len`... Al llamarlas, casi siempre queremos *hacer algo* con el resultado: asignarlo a una variable o usarlo en una expresión.
- Las **funciones vacías** (*void*), como `print_twice`, realizan una acción pero **no devuelven valor**.

⚠️ En un **script**, si se llama una función fructífera sin guardar su resultado, ¡el valor de retorno **se desvanece**! (`math.sqrt(5)` calcula la raíz de 5, pero sin asignación ni impresión el resultado se pierde).

Si se intenta asignar el resultado de una función vacía a una variable, se obtiene el valor especial **`None`**. `None` **no** es la cadena `"None"`: es un valor especial con su propio tipo, `NoneType`.

Para devolver un resultado desde una función se usa la sentencia **`return`**. Por ejemplo, `addtwo(a, b)` suma dos números y devuelve el resultado: los parámetros `a` y `b` reciben los argumentos, la suma se guarda en la variable local `added` y `return` envía ese valor de vuelta al código llamador.

````python
import math

# Función FRUCTÍFERA: el resultado se asigna o se usa en una expresión
x = math.cos(0.7)
golden = (math.sqrt(5) + 1) / 2
print(f'cos(0.7) = {x}')
print(f'Número áureo = {golden}')

# Función VACÍA (void): realiza una acción pero devuelve None
def print_twice(bruce):
    print(bruce)
    print(bruce)

result = print_twice('Bing')   # imprime Bing dos veces...
print(result)                  # ...pero result es None
print(type(None))              # <class 'NoneType'>
````

````python
# La sentencia RETURN devuelve un resultado al código llamador
def addtwo(a, b):
    added = a + b      # variable LOCAL a la función
    return added       # envía el valor calculado de vuelta

x = addtwo(3, 5)       # a=3, b=5 -> added=8 -> return 8
print(x)               # 8

# El valor de retorno puede usarse directamente en expresiones
print(addtwo(10, 20) * 2)   # 60
````

## 10. ¿Por qué usar funciones?

Puede no ser evidente por qué vale la pena dividir un programa en funciones. Hay varias razones:

1. **Legibilidad:** crear una función permite *nombrar* un grupo de sentencias, lo que hace el programa más fácil de leer, entender y depurar.
2. **Eliminación de código repetitivo:** las funciones hacen el programa más pequeño. Si luego hay que hacer un cambio, solo se modifica **en un lugar**.
3. **Depuración incremental:** dividir un programa largo en funciones permite depurar las partes una por una y luego ensamblarlas en un todo funcional.
4. **Reutilización:** las funciones bien diseñadas sirven para muchos programas. Una vez escrita y depurada, se puede reutilizar.

A lo largo del curso usaremos definiciones de función para explicar conceptos. Parte de la habilidad consiste en lograr que una función **capture correctamente una idea**, como *"encontrar el valor más pequeño de una lista"*.

## 12. Notas para buenas prácticas 

### a) *Docstrings* — documentación integrada
Cadena de texto como primera sentencia del cuerpo; describe qué hace la función, sus parámetros y su valor de retorno. Es accesible en tiempo de ejecución con `help(funcion)` y es leída por generadores de documentación (Sphinx, pydoc).

### b) Anotaciones de tipos (*type hints*, PEP 484)
`def area(radio: float) -> float:` declara los tipos esperados. No son obligatorias en tiempo de ejecución, pero permiten que herramientas de análisis estático (Pylance, mypy) detecten errores **antes** de ejecutar.

### c) Argumentos por defecto y por nombre
Los parámetros pueden tener valores por defecto (`def f(x, escala=1.0)`) y las llamadas pueden nombrar los argumentos (`f(3, escala=2.5)`), lo que mejora la legibilidad.

### d) `*args` 
Permiten funciones con un número variable de argumentos posicionales (`*args`).

### e) Funciones `lambda`
Funciones anónimas de una sola expresión, útiles como argumentos de orden superior (`sorted`, `map`, `filter`).

### Ejercicio 6: Cálculo de pago con horas extra

Reescriba el cálculo del salario con **tiempo y medio** para las horas extra, creando una función llamada `computepay` que reciba dos parámetros (`hours` y `rate`).

**Regla:** las primeras 40 horas se pagan a la tarifa normal; cada hora adicional se paga a 1.5 veces la tarifa.

**Comportamiento esperado:**

```
Enter Hours: 45
Enter Rate: 10
Pay: 475.0
```

Verificación: $40 \times 10 + 5 \times (10 \times 1.5) = 400 + 75 = 475$

> ✍️ Intente escribir su propia solución antes de consultar la celda de solución.

````python
# SOLUCIÓN Ejercicio 6: computepay con tiempo y medio para horas extra
def computepay(hours, rate):
    if hours > 40:
        pay = 40 * rate + (hours - 40) * rate * 1.5
    else:
        pay = hours * rate
    return pay

# Entrada de datos
hours = float(input('Enter Hours: '))
rate = float(input('Enter Rate: '))
print('Pay:', computepay(hours, rate))

# Pruebas rápidas: 45 horas a tarifa 10 -> 475.0
print(computepay(45, 10))   # 475.0
print(computepay(40, 10))   # 400.0
print(computepay(30, 10))   # 300.0
````

### Ejercicio 7: Cálculo de calificaciones

Reescriba el programa de calificaciones del capítulo anterior usando una función llamada `computegrade` que reciba un puntaje (*score*) como parámetro y **devuelva la calificación como cadena**.

| Puntaje  | Calificación |
|----------|:---:|
| >= 0.9   | A |
| >= 0.8   | B |
| >= 0.7   | C |
| >= 0.6   | D |
| < 0.6    | F |

**Comportamiento esperado** (ejecute el programa varias veces para probar distintos valores):

```
Enter score: 0.95      ->  A
Enter score: perfect   ->  Bad score
Enter score: 10.0      ->  Bad score
Enter score: 0.75      ->  C
Enter score: 0.5       ->  F
```

**Requisitos de la solución:**
1. La entrada no numérica debe capturarse con `try/except` y mostrar `Bad score`.
2. Los puntajes fuera del rango [0.0, 1.0] también deben producir `Bad score`.
3. La función debe **devolver** la calificación (función fructífera), no solo imprimirla.

````python
# SOLUCIÓN Ejercicio 7: computegrade devuelve la calificación como cadena
def computegrade(score: float) -> str:
    """
    Convierte un puntaje numérico [0.0, 1.0] en calificación alfabética.

    Parámetros:
        score (float): puntaje entre 0.0 y 1.0.

    Retorna:
        str: 'A', 'B', 'C', 'D' o 'F'; 'Bad score' si está fuera de rango.
    """
    if score < 0.0 or score > 1.0:
        return 'Bad score'
    elif score >= 0.9:
        return 'A'
    elif score >= 0.8:
        return 'B'
    elif score >= 0.7:
        return 'C'
    elif score >= 0.6:
        return 'D'
    else:
        return 'F'

# Entrada con manejo de excepciones para valores no numéricos
try:
    score = float(input('Enter score: '))
    print(computegrade(score))
except ValueError:
    print('Bad score')

# Pruebas con distintos valores (incluye los casos del enunciado)
casos = [0.95, 0.85, 0.75, 0.65, 0.5, 10.0, -0.1]
for s in casos:
    print(f'score = {s:>5} -> {computegrade(s)}')

computegrade(0.95)
computegrade(-0.1)
print('Pruebas de computegrade superadas ✔')
````
