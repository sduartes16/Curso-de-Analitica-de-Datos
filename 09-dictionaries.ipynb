Un *diccionario* es como una lista, pero más general. En una lista, las posiciones de los índices tienen que ser enteros; en un diccionario, los índices pueden ser (casi) de cualquier tipo.

Puedes pensar en un diccionario como un mapeo entre un conjunto de índices (que se llaman *claves*, keys) y un conjunto de valores. Cada clave mapea a un valor. La asociación de una clave y un valor se llama un *par clave-valor* o a veces un *ítem*.

Como ejemplo, construiremos un diccionario que mapea de palabras en inglés a palabras en español, así que las claves y los valores son todos cadenas.

La función `dict` crea un nuevo diccionario sin ítems. Como `dict` es el nombre de una función incorporada, deberías evitar usarlo como nombre de variable.


````python
eng2sp = dict()
print(eng2sp)   # {}
````

Las llaves, `{}`, representan un diccionario vacío. Para añadir ítems al diccionario, puedes usar corchetes:


````python
eng2sp['one'] = 'uno' 
````

Esta línea crea un ítem que mapea de la clave `'one'` al valor "uno". Si imprimimos el diccionario de nuevo, vemos un par clave-valor con dos puntos entre la clave y el valor:


````python
print(eng2sp)   # {'one': 'uno'}
````

Este formato de salida es también un formato de entrada. Por ejemplo, puedes crear un nuevo diccionario con tres ítems.


````python
eng2sp = {'one': 'uno', 'two': 'dos', 'three': 'tres'}
print(eng2sp)   # {'one': 'uno', 'two': 'dos', 'three': 'tres'}
````

Desde Python 3.7 el orden de los pares clave-valor es el mismo que su orden de entrada, es decir, los diccionarios son ahora estructuras ordenadas.

Pero eso en realidad no importa mucho porque los elementos de un diccionario nunca se indexan con índices enteros. En su lugar, usas las claves para buscar los valores correspondientes:


````python
print(eng2sp['two'])   # dos
````

La clave `'two'` siempre mapea al valor "dos", así que el orden de los ítems no importa.

Si la clave no está en el diccionario, obtienes una excepción:

```
>>> print(eng2sp['four'])
KeyError: 'four'
```

La función `len` funciona con diccionarios; devuelve el número de pares clave-valor:


````python
len(eng2sp)   # 3
````

El operador `in` funciona con diccionarios; te dice si algo aparece como *clave* en el diccionario (aparecer como valor no es suficiente).


````python
eng2sp = {'one': 'uno', 'two': 'dos', 'three': 'tres'}
'one' in eng2sp   # True
'uno' in eng2sp   # False
````

Para ver si algo aparece como un valor en un diccionario, puedes usar el método `values`, que devuelve los valores como un tipo que puede convertirse a una lista, y luego usar el operador `in`:


````python
vals = list(eng2sp.values())
'uno' in vals   # True
````

El operador `in` usa algoritmos diferentes para las listas y los diccionarios. Para las listas, usa un algoritmo de búsqueda lineal. A medida que la lista se hace más larga, el tiempo de búsqueda crece en proporción directa a la longitud de la lista. Para los diccionarios, Python usa un algoritmo llamado *tabla hash* que tiene una propiedad notable: el operador `in` tarda aproximadamente el mismo tiempo sin importar cuántos ítems haya en el diccionario. No explicaré por qué las funciones hash son tan mágicas, pero puedes leer más sobre ello en [wikipedia.org/wiki/Hash_table](https://wikipedia.org/wiki/Hash_table). *(Nota: si quieres aprender más sobre tablas hash, hay un curso en https://www.cc4e.com que explora cómo el lenguaje de programación C implementa un diccionario de Python.)*

## El diccionario como un conjunto de contadores

Supón que te dan una cadena y quieres contar cuántas veces aparece cada letra. Hay varias formas de hacerlo:

1. Podrías crear 26 variables, una para cada letra del alfabeto. Luego podrías recorrer la cadena y, para cada carácter, incrementar el contador correspondiente, probablemente usando un condicional encadenado.

2. Podrías crear una lista con 26 elementos. Luego podrías convertir cada carácter a un número (usando la función incorporada `ord`), usar el número como índice de la lista e incrementar el contador apropiado.

3. Podrías crear un diccionario con los caracteres como claves y los contadores como los valores correspondientes. La primera vez que vieras un carácter, añadirías un ítem al diccionario. Después de eso, incrementarías el valor del ítem existente.

Cada una de estas opciones realiza el mismo cálculo, pero cada una implementa ese cálculo de una manera diferente.

Una *implementación* es una forma de realizar un cálculo; algunas implementaciones son mejores que otras. Por ejemplo, una ventaja de la implementación con diccionario es que no tenemos que saber por adelantado qué letras aparecen en la cadena, y solo tenemos que hacer espacio para las letras que sí aparecen.

Así es como podría verse el código:


````python
word = 'brontosaurus'
d = dict()
for c in word:
    if c not in d:
        d[c] = 1
    else:
        d[c] = d[c] + 1
d
````

Efectivamente estamos calculando un *histograma*, que es un término estadístico para un conjunto de contadores (o frecuencias).

El bucle `for` recorre la cadena. Cada vez que pasa por el bucle, si `c` no está en el diccionario, creamos un nuevo ítem con clave `c` y valor inicial 1 (ya que hemos visto esta letra una vez). Si `c` ya está en el diccionario, incrementamos `d[c]`.

Esta es la salida del programa:

```
{'b': 1, 'r': 2, 'o': 2, 'n': 1, 't': 1, 's': 2, 'a': 1, 'u': 2}
```

El histograma indica que las letras "a" y "b" aparecen una vez; "o" aparece dos veces, y así sucesivamente.

Los diccionarios tienen un método llamado `get` que toma una clave y un valor por defecto. Si la clave aparece en el diccionario, `get` devuelve el valor correspondiente; de lo contrario devuelve el valor por defecto. Por ejemplo:


````python
counts = { 'chuck' : 1 , 'annie' : 42, 'jan': 100}
counts.get('jan', 0)   # 100
#counts.get('tim', 0)   # 0
````

Podemos usar `get` para escribir nuestro bucle del histograma de forma más concisa. Como el método `get` maneja automáticamente el caso en que una clave no está en el diccionario, podemos reducir cuatro líneas a una y eliminar la sentencia `if`.


````python
word = 'brontosaurus'
d = dict()
for c in word:
    d[c] = d.get(c,0) + 1
d
````

El uso del método `get` para simplificar este bucle de conteo termina siendo un "idioma" (idiom) muy usado en Python y lo usaremos muchas veces en el resto del libro. Así que deberías tomarte un momento y comparar el bucle que usa la sentencia `if` y el operador `in` con el bucle que usa el método `get`. Hacen exactamente lo mismo, pero uno es más sucinto.


## Diccionarios y archivos

Uno de los usos comunes de un diccionario es contar la ocurrencia de palabras en un archivo con algún texto escrito. Empecemos con un archivo de palabras muy simple tomado del texto de *Romeo y Julieta*.

Para el primer conjunto de ejemplos, usaremos una versión acortada y simplificada del texto sin puntuación. Más adelante trabajaremos con el texto de la escena con la puntuación incluida.

    But soft what light through yonder window breaks
    It is the east and Juliet is the sun
    Arise fair sun and kill the envious moon
    Who is already sick and pale with grief

Escribiremos un programa de Python para leer las líneas del archivo, dividir cada línea en una lista de palabras, y luego recorrer cada una de las palabras de la línea y contar cada palabra usando un diccionario.

Verás que tenemos dos bucles `for`. El bucle externo está leyendo las líneas del archivo y el bucle interno está iterando a través de cada una de las palabras de esa línea en particular. Este es un ejemplo de un patrón llamado *bucles anidados* porque uno de los bucles es el bucle *externo* y el otro bucle es el bucle *interno*.

Como el bucle interno ejecuta todas sus iteraciones cada vez que el bucle externo hace una sola iteración, pensamos que el bucle interno itera "más rápido" y el bucle externo itera más lentamente.

La combinación de los dos bucles anidados asegura que contaremos cada palabra de cada línea del archivo de entrada. Este es el programa `count1.py`:


````python
fname = input('Enter the file name: ')
try:
    fhand = open(fname)
except:
    print('File cannot be opened:', fname)
    exit()

counts = dict()
for line in fhand:
    words = line.split()
    for word in words:
        if word not in counts:
            counts[word] = 1
        else:
            counts[word] += 1

counts
````

*(El programa lee el archivo `romeo.txt`, que está junto a este notebook.)*

En nuestra sentencia `else`, usamos la alternativa más compacta para incrementar una variable. `counts[word] += 1` es equivalente a `counts[word] = counts[word] + 1`. Cualquiera de los dos métodos puede usarse para cambiar el valor de una variable en cualquier cantidad deseada. Existen alternativas similares para `-=`, `*=` y `/=`.

Cuando ejecutamos el programa, vemos un volcado en bruto de todos los conteos en orden hash sin ordenar. (el archivo *romeo.txt* está disponible en [www.py4e.com/code3/romeo.txt](http://www.py4e.com/code3/romeo.txt))

## Bucles y diccionarios

Si usas un diccionario como la secuencia en una sentencia `for`, éste recorre las claves del diccionario. Este bucle imprime cada clave y el valor correspondiente:


````python
counts = { 'chuck' : 1 , 'annie' : 42, 'jan': 100}
for key in counts:
    print(key, counts[key])

````

Así es como se ve la salida:

```
chuck 1
annie 42
jan 100
```

De nuevo, las claves están ordenadas.

Podemos usar este patrón para implementar los diversos idiomas de bucle que hemos descrito antes. Por ejemplo, si quisiéramos encontrar todas las entradas de un diccionario con un valor mayor que diez, podríamos escribir el siguiente código:


````python
counts = { 'chuck' : 1 , 'annie' : 42, 'jan': 100}
for key in counts:
    if counts[key] > 10 :
        print(key, counts[key])
````

El bucle `for` itera a través de las *claves* del diccionario, así que debemos usar el operador de índice para recuperar el *valor* correspondiente a cada clave. Así es como se ve la salida:

```
annie 42
jan 100
```

Vemos solo las entradas con un valor por encima de 10.

Si quieres imprimir las claves en orden alfabético, primero haces una lista de las claves del diccionario usando el método `keys` disponible en los objetos diccionario, y luego ordenas esa lista y recorres la lista ordenada, buscando cada clave e imprimiendo los pares clave-valor en orden, de la siguiente manera:


````python
counts = { 'chuck' : 1 , 'annie' : 42, 'jan': 100}
lst = list(counts.keys())
print(lst)
lst.sort()
print(lst)
for key in lst:
    print(key, counts[key])
````

Así es como se ve la salida:

```
['chuck', 'annie', 'jan']
['annie', 'chuck', 'jan']
annie 42
chuck 1
jan 100
```

Primero ves la lista de claves en orden no alfabético que obtenemos del método `keys`. Luego vemos los pares clave-valor en orden alfabético del bucle `for`.


## Análisis de texto avanzado

En el ejemplo anterior que usaba el archivo *romeo.txt*, hicimos el archivo lo más simple posible eliminando a mano toda la puntuación. El texto real tiene mucha puntuación, como se muestra a continuación.

```
But, soft! what light through yonder window breaks?
It is the east, and Juliet is the sun.
Arise, fair sun, and kill the envious moon,
Who is already sick and pale with grief,
```

Dado que la función `split` de Python busca espacios y trata las palabras como tokens separados por espacios, trataríamos las palabras "soft!" y "soft" como palabras *diferentes* y crearíamos una entrada de diccionario separada para cada palabra.

Además, como el archivo tiene mayúsculas, trataríamos "who" y "Who" como palabras diferentes con conteos diferentes.

Podemos resolver ambos problemas usando los métodos de cadenas `lower`, `punctuation` y `translate`. El método `translate` es el más sutil de los métodos. Aquí está la documentación de `translate`:

`line.translate(str.maketrans(fromstr, tostr, deletestr))`

*Reemplaza los caracteres de `fromstr` con el carácter en la misma posición de `tostr` y elimina todos los caracteres que están en `deletestr`. `fromstr` y `tostr` pueden ser cadenas vacías y el parámetro `deletestr` puede omitirse.*

No especificaremos el `tostr` pero usaremos el parámetro `deletestr` para eliminar toda la puntuación. Incluso dejaremos que Python nos diga la lista de caracteres que considera "puntuación":


````python
import string
string.punctuation   # '!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~' 
````

Los parámetros usados por `translate` eran diferentes en Python 2.0.

Hacemos las siguientes modificaciones a nuestro programa (`count2.py`):


````python
import string

fname = input('Enter the file name: ')
try:
    fhand = open(fname)
except:
    print('File cannot be opened:', fname)
    exit()

counts = dict()
for line in fhand:
    line = line.rstrip()
    # First two parameters are empty strings
    line = line.translate(line.maketrans("", "", string.punctuation))
    line = line.lower()
    words = line.split()
    for word in words:
        if word not in counts:
            counts[word] = 1
        else:
            counts[word] += 1

print(counts)
````

## Depuración

A medida que trabajas con conjuntos de datos más grandes, puede volverse difícil depurar imprimiendo y comprobando los datos a mano. Aquí hay algunas sugerencias para depurar grandes conjuntos de datos:

**Reducir la escala de la entrada**: Si es posible, reduce el tamaño del conjunto de datos. Por ejemplo, si el programa lee un archivo de texto, empieza solo con las primeras 10 líneas, o con el ejemplo más pequeño que puedas encontrar. Puedes editar los archivos directamente, o (mejor) modificar el programa para que lea solo las primeras `n` líneas.

Si hay un error, puedes reducir `n` al valor más pequeño que manifieste el error, y luego aumentarlo gradualmente a medida que encuentras y corriges errores.

**Comprobar resúmenes y tipos**: En lugar de imprimir y comprobar el conjunto de datos completo, considera imprimir resúmenes de los datos: por ejemplo, el número de ítems de un diccionario o el total de una lista de números.

Una causa común de errores en tiempo de ejecución es un valor que no es del tipo correcto. Para depurar este tipo de error, a menudo basta con imprimir el tipo de un valor.

**Escribir autocomprobaciones**: A veces puedes escribir código para comprobar errores automáticamente. Por ejemplo, si estás calculando el promedio de una lista de números, podrías comprobar que el resultado no es mayor que el elemento más grande de la lista ni menor que el más pequeño. Esto se llama una "comprobación de cordura" (sanity check) porque detecta resultados que son "completamente ilógicos".

Otro tipo de comprobación compara los resultados de dos cálculos diferentes para ver si son consistentes. Esto se llama una "comprobación de consistencia".

**Imprimir la salida con formato**: Formatear la salida de depuración puede hacer más fácil detectar un error.

De nuevo, el tiempo que dediques a construir andamiaje (scaffolding) puede reducir el tiempo que dediques a depurar.


## Glosario

- **diccionario (dictionary)**: Un mapeo de un conjunto de claves a sus valores correspondientes.

- **tabla hash (hashtable)**: El algoritmo usado para implementar los diccionarios de Python.

- **función hash (hash function)**: Una función usada por una tabla hash para calcular la ubicación de una clave.

- **histograma (histogram)**: Un conjunto de contadores.

- **implementación (implementation)**: Una forma de realizar un cálculo.

- **ítem (item)**: Otro nombre para un par clave-valor.

- **clave (key)**: Un objeto que aparece en un diccionario como la primera parte de un par clave-valor.

- **par clave-valor (key-value pair)**: La representación del mapeo de una clave a un valor.

- **búsqueda (lookup)**: Una operación de diccionario que toma una clave y encuentra el valor correspondiente.

- **bucles anidados (nested loops)**: Cuando hay uno o más bucles "dentro" de otro bucle. El bucle interno se ejecuta hasta completarse cada vez que el bucle externo se ejecuta una vez.

- **valor (value)**: Un objeto que aparece en un diccionario como la segunda parte de un par clave-valor. Esto es más específico que nuestro uso anterior de la palabra "valor".
