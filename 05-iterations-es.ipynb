## Ideas clave

- Una variable de control debe inicializarse antes del bucle.
- El cuerpo de un `while` debe cambiar esa variable para que el bucle termine.
- `break` termina el bucle inmediatamente.
- `continue` salta al comienzo de la siguiente iteración.
- `for` recorre cada elemento de una secuencia.
- Un contador cuenta elementos; un acumulador conserva un total.

## Bucle `while`: repetir mientras una condición sea verdadera

La variable `n` disminuye en cada vuelta. Cuando llega a cero, la condición `n > 0` deja de cumplirse.

```python
n = 5
while n > 0:
    print(n)
    n = n - 1
print('¡Despegue!')
```

## `while True` con `break`

Este patrón sirve cuando la condición de salida se comprueba dentro del bucle. La palabra `done` termina la ejecución.

```python
while True:
    texto = input('Escribe algo (done para terminar): ')
    if texto == 'done':
        break
    print('Escribiste:', texto)
print('Programa terminado')
```

## `continue`: saltar una iteración

Las líneas que empiezan con `#` se ignoran. El `continue` evita imprimirlas y vuelve al inicio del bucle.

```python
valores = [3, -1, 5, -2, 8]
for valor in valores:
    if valor < 0:
        continue
    print('Valor positivo:', valor)
```

## Bucle `for`: recorrer una lista

`for` ejecuta el cuerpo una vez por cada elemento de la lista.

```python
amigos = ['Joseph', 'Glenn', 'Sally']
for amigo in amigos:
    print('Feliz año nuevo:', amigo)
print('Terminado')
```

## Contador y acumulador

Inicializa las variables antes del bucle. El contador aumenta una vez por elemento y el acumulador suma el valor de cada elemento.

```python
valores = [3, 41, 12, 9, 74, 15]
contador = 0
total = 0

for valor in valores:
    contador = contador + 1
    total = total + valor

print('Cantidad:', contador)
print('Total:', total)
print('Promedio:', total / contador)
```

## Máximo y mínimo con `None`

`None` marca que todavía no hemos visto ningún valor. La condición `or` permite guardar el primer elemento y luego comparar los siguientes.

```python
valores = [3, 41, 12, 9, 74, 15]
mayor = None
menor = None

for valor in valores:
    if mayor is None or valor > mayor:
        mayor = valor
    if menor is None or valor < menor:
        menor = valor

print('Mayor:', mayor)
print('Menor:', menor)
```
