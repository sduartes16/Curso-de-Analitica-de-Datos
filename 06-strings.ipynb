## Una cadena es una secuencia

Una cadena es una *secuencia* de caracteres. Puedes acceder a los caracteres de uno en uno con el operador de corchetes:


````python
fruit = 'banana'
letter = fruit[1]
````

La segunda sentencia extrae el carácter en la posición de índice 1 de la variable `fruit` y lo asigna a la variable `letter`.

La expresión entre corchetes se llama un *índice*. El índice indica qué carácter de la secuencia quieres (de ahí el nombre).

Pero puede que no obtengas lo que esperas:


````python
print(letter)   # a
````

Para la mayoría de la gente, la primera letra de "banana" es "b", no "a". Pero en Python, el índice es un desplazamiento desde el comienzo de la cadena, y el desplazamiento de la primera letra es cero.


````python
letter = fruit[0]
print(letter)   # b
````

Así que "b" es la letra 0-ésima ("cero-ésima") de "banana", "a" es la letra 1-ésima ("uno-ésima"), y "n" es la letra 2-ésima ("dos-ésima").


Puedes usar cualquier expresión, incluyendo variables y operadores, como índice, pero el valor del índice tiene que ser un entero. De lo contrario obtienes:

```
>>> letter = fruit[1.5]
TypeError: string indices must be integers
```


## Obtener la longitud de una cadena usando `len`

`len` es una función incorporada que devuelve el número de caracteres de una cadena:


````python
fruit = 'banana'
len(fruit)   # 6
````

Para obtener la última letra de una cadena, podrías sentirte tentado a intentar algo como esto:

```
>>> length = len(fruit)
>>> last = fruit[length]
IndexError: string index out of range
```

La razón del `IndexError` es que no hay ninguna letra en "banana" con el índice 6. Como empezamos a contar en cero, las seis letras están numeradas del 0 al 5. Para obtener el último carácter, tienes que restar 1 a `length`:


````python
length = len(fruit)
last = fruit[length-1]
print(last)   # a
````

Alternativamente, puedes usar índices negativos, que cuentan hacia atrás desde el final de la cadena. La expresión `fruit[-1]` produce la última letra, `fruit[-2]` produce la penúltima, y así sucesivamente.


## Recorrido de una cadena con un bucle

Muchos cálculos implican procesar una cadena carácter a carácter. A menudo empiezan por el principio, seleccionan cada carácter por turno, hacen algo con él y continúan hasta el final. Este patrón de procesamiento se llama un *recorrido* (traversal). Una forma de escribir un recorrido es con un bucle `while`:


````python
index = 0
while index < len(fruit):
    letter = fruit[index]
    print(letter)
    index = index + 1
````

Este bucle recorre la cadena y muestra cada letra en una línea propia. La condición del bucle es `index < len(fruit)`, así que cuando `index` es igual a la longitud de la cadena, la condición es falsa y el cuerpo del bucle no se ejecuta. El último carácter accedido es el que tiene el índice `len(fruit)-1`, que es el último carácter de la cadena.

## Rebanadas de cadenas (slices)

Un segmento de una cadena se llama una *rebanada* (slice). Seleccionar una rebanada es similar a seleccionar un carácter:


````python
s = 'Monty Python'
print(s[0:5])    # Monty
print(s[6:12])   # Python
````

El operador [n:m] devuelve la parte de la cadena desde el carácter "n-ésimo" hasta el carácter "m-ésimo", incluyendo el primero pero excluyendo el último.

Si omites el primer índice (antes de los dos puntos), la rebanada comienza al principio de la cadena. Si omites el segundo índice, la rebanada llega hasta el final de la cadena:


````python
fruit = 'banana'
fruit[:3]   # 'ban'
fruit[3:]   # 'ana'
````

Si el primer índice es mayor o igual que el segundo, el resultado es una *cadena vacía*, representada por dos comillas:


````python
fruit = 'banana'
fruit[3:3]   # '' 
````

Una cadena vacía no contiene caracteres y tiene longitud 0, pero aparte de eso, es igual que cualquier otra cadena.

## Las cadenas son inmutables

Es tentador usar el operador `[]` en el lado izquierdo de una asignación, con la intención de cambiar un carácter de una cadena. Por ejemplo:

```
>>> greeting = 'Hello, world!'
>>> greeting[0] = 'J'
TypeError: 'str' object does not support item assignment
```

El "objeto" en este caso es la cadena y el "ítem" es el carácter que intentaste asignar. Por ahora, un *objeto* es lo mismo que un valor, pero refinaremos esa definición más adelante. Un *ítem* es uno de los valores de una secuencia.

La razón del error es que las cadenas son *inmutables*, lo que significa que no puedes cambiar una cadena existente. Lo mejor que puedes hacer es crear una nueva cadena que sea una variación de la original:


````python
greeting = 'Hello, world!'
new_greeting = 'J' + greeting[1:]
print(new_greeting)   # Jello, world!
````

Este ejemplo concatena una nueva primera letra con una rebanada de `greeting`. No tiene ningún efecto sobre la cadena original.


## Bucles y conteo

El siguiente programa cuenta el número de veces que la letra "a" aparece en una cadena:


````python
word = 'banana'
count = 0
for letter in word:
    if letter == 'a':
        count = count + 1
print(count)
````

Este programa demuestra otro patrón de computación llamado *contador*. La variable `count` se inicializa a 0 y luego se incrementa cada vez que se encuentra una "a". Cuando el bucle termina, `count` contiene el resultado: el número total de letras "a".

## El operador `in`

La palabra `in` es un operador booleano que toma dos cadenas y devuelve `True` si la primera aparece como subcadena en la segunda:


````python
fruit = 'banana'
'a' in 'banana'      # True
'seed' in 'banana'   # False
````

## Comparación de cadenas

Los operadores de comparación funcionan con cadenas. Para ver si dos cadenas son iguales:


````python
word = 'banana'   # cambia la palabra y vuelve a ejecutar

if word == 'banana':
    print('All right, bananas.')
````

Otras operaciones de comparación son útiles para poner palabras en orden alfabético:


````python
word = 'Pineapple'   # prueba con otras palabras y vuelve a ejecutar

if word < 'banana':
    print('Your word,' + word + ', comes before banana.')
elif word > 'banana':
    print('Your word,' + word + ', comes after banana.')
else:
    print('All right, bananas.')
````

Python no maneja las letras mayúsculas y minúsculas de la misma manera que las personas. Todas las letras mayúsculas van antes que todas las letras minúsculas, así que:

```
Your word, Pineapple, comes before banana.
```

Una forma común de abordar este problema es convertir las cadenas a un formato estándar, como todo en minúsculas, antes de realizar la comparación. Tenlo en cuenta por si alguna vez tienes que defenderte de un hombre armado con una Piña.


## Métodos de las cadenas

Las cadenas son un ejemplo de *objetos* de Python. Un objeto contiene tanto datos (la cadena en sí) como *métodos*, que son efectivamente funciones incorporadas en el objeto y disponibles para cualquier *instancia* del objeto.

Python tiene una función llamada `dir` que lista los métodos disponibles para un objeto. La función `type` muestra el tipo de un objeto y la función `dir` muestra los métodos disponibles.


````python
stuff = 'Hello world'
type(stuff)   # <class 'str'>
````

````python
dir(stuff)
````

El resultado incluye métodos como 'capitalize', 'casefold', 'center', 'count', 'encode', 'endswith', 'expandtabs', 'find', 'format', 'format_map', 'index', 'isalnum', 'isalpha', 'isdecimal', 'isdigit', 'isidentifier', 'islower', 'isnumeric', 'isprintable', 'isspace', 'istitle', 'isupper', 'join', 'ljust', 'lower', 'lstrip', 'maketrans', 'partition', 'replace', 'rfind', 'rindex', 'rjust', 'rpartition', 'rsplit', 'rstrip', 'split', 'splitlines', 'startswith', 'strip', 'swapcase', 'title', 'translate', 'upper' y 'zfill'.


````python
help(str.capitalize)
````

Mientras que la función `dir` lista los métodos, y puedes usar `help` para obtener documentación sencilla sobre un método, una mejor fuente de documentación para los métodos de las cadenas sería:

<https://docs.python.org/library/stdtypes.html#string-methods>

Llamar a un *método* es similar a llamar a una función (toma argumentos y devuelve un valor) pero la sintaxis es diferente. Llamamos a un método añadiendo el nombre del método al nombre de la variable usando el punto como delimitador.

Por ejemplo, el método `upper` toma una cadena y devuelve una nueva cadena con todas las letras en mayúsculas:

En lugar de la sintaxis de función `upper(word)`, usa la sintaxis de método `word.upper()`.


````python
word = 'banana'
new_word = word.upper()
print(new_word)   # BANANA
````

Esta forma de notación de punto especifica el nombre del método, `upper`, y el nombre de la cadena a la que aplicar el método, `word`. Los paréntesis vacíos indican que este método no toma ningún argumento.

Una llamada a un método se llama una *invocación*; en este caso, diríamos que estamos invocando `upper` sobre `word`.

Por ejemplo, hay un método de las cadenas llamado `find` que busca la posición de una cadena dentro de otra:


````python
word = 'banana'
index = word.find('a')
print(index)   # 1
````

En este ejemplo, invocamos `find` sobre `word` y pasamos la letra que estamos buscando como parámetro.

El método `find` puede encontrar subcadenas además de caracteres:


````python
word.find('na')   # 2
````

Puede tomar como segundo argumento el índice donde debe empezar:


````python
word.find('na', 3)   # 4
````

Una tarea común es eliminar los espacios en blanco (espacios, tabuladores o saltos de línea) del principio y del final de una cadena usando el método `strip`:


````python
line = '  Here we go  '
line.strip()   # 'Here we go'
````

Algunos métodos como *startswith* devuelven valores booleanos.


````python
line = 'Have a nice day'
line.startswith('Have')   # True
line.startswith('h')      # False
````

Notarás que `startswith` requiere que coincidan las mayúsculas y minúsculas, así que a veces tomamos una línea y la pasamos toda a minúsculas antes de hacer cualquier comprobación usando el método `lower`.


````python
line = 'Have a nice day'
line.startswith('h')            # False
line.lower()                    # 'have a nice day'
line.lower().startswith('h')    # True
````

En el último ejemplo, se llama al método `lower` y luego usamos `startswith` para ver si la cadena resultante en minúsculas comienza con la letra "h". Siempre que seamos cuidadosos con el orden, podemos hacer múltiples llamadas a métodos en una sola expresión.

## Análisis de cadenas (parsing)

A menudo, queremos mirar dentro de una cadena y encontrar una subcadena. Por ejemplo, si se nos presentara una serie de líneas con el siguiente formato:

`From stephen.marquard@`*`uct.ac.za`*` Sat Jan  5 09:14:16 2008`

y quisiéramos extraer solo la segunda mitad de la dirección (es decir, `uct.ac.za`) de cada línea, podemos hacerlo usando el método `find` y las rebanadas de cadenas.

Primero, encontraremos la posición de la arroba en la cadena. Luego encontraremos la posición del primer espacio *después* de la arroba. Y luego usaremos las rebanadas de cadena para extraer la porción de la cadena que estamos buscando.


````python
data = 'From stephen.marquard@uct.ac.za Sat Jan  5 09:14:16 2008'
atpos = data.find('@')
print(atpos)          # 21
sppos = data.find(' ',atpos)
print(sppos)          # 31
host = data[atpos+1:sppos]
print(host)           # uct.ac.za
````

Usamos una versión del método `find` que nos permite especificar una posición en la cadena donde queremos que `find` empiece a buscar. Al hacer la rebanada, extraemos los caracteres desde "uno más allá de la arroba hasta el carácter de espacio, *pero sin incluirlo*".

## Literales de cadena formateados (f-strings)

Un literal de cadena formateado (al que a menudo se hace referencia simplemente como f-string) permite usar expresiones de Python dentro de literales de cadena. Esto se logra anteponiendo una `f` al literal de cadena y encerrando las expresiones entre llaves `{}`.

Por ejemplo, envolver el nombre de una variable entre llaves dentro de un f-string hará que sea reemplazada por su valor:


````python
camels = 42
f'{camels}'   # '42'
````

El resultado es la cadena '42', que no debe confundirse con el valor entero 42.

Una expresión puede aparecer en cualquier parte de la cadena, así que puedes incrustar un valor en una oración:


````python
camels = 42
f'I have spotted {camels} camels.'   # 'I have spotted 42 camels.'
````

Se pueden incluir varias expresiones dentro de un solo literal de cadena para crear cadenas más complejas.


````python
years = 3
count = .1
species = 'camels'
f'In {years} years I have spotted {count} {species}.'   # 'In 3 years I have spotted 0.1 camels.'
````

Los literales de cadena formateados son potentes, y pueden hacer incluso más de lo que se cubre aquí. Puedes leer más sobre ellos en:

<https://docs.python.org/3/tutorial/inputoutput.html#formatted-string-literals>


## Depuración

Una habilidad que deberías cultivar mientras programas es preguntarte siempre: "¿Qué podría salir mal aquí?" o alternativamente: "¿Qué locura podría hacer nuestro usuario para hacer fallar nuestro (aparentemente) perfecto programa?"

Por ejemplo, mira el programa que usamos para demostrar el bucle `while` en el capítulo sobre iteración (`copytildone2.py`):


````python
while True:
    line = input('> ')
    if line[0] == '#':
        continue
    if line == 'done':
        break
    print(line)
print('Done!')
````

Mira lo que ocurre cuando el usuario introduce una línea vacía como entrada:

```
> hello there
hello there
> # don't print this
> print this!
print this!
>
Traceback (most recent call last):
  File "copytildone.py", line 3, in <module>
    if line[0] == '#':
IndexError: string index out of range
```

El código funciona bien hasta que se le presenta una línea vacía. Entonces no hay carácter cero-ésimo, así que obtenemos un traceback. Hay dos soluciones para esto, para hacer que la línea tres sea "segura" incluso si la línea está vacía.

Una posibilidad es simplemente usar el método `startswith`, que devuelve `False` si la cadena está vacía.


````python
line = ''   # prueba también con '# comentario' o 'texto'

if line.startswith('#'):
    print('Es un comentario')
````

Otra forma es escribir la sentencia `if` de forma segura usando el patrón *guardián* (guardian pattern) y asegurarse de que la segunda expresión lógica solo se evalúa cuando hay al menos un carácter en la cadena:


````python
line = ''   # prueba también con '# comentario' o 'texto'

if len(line) > 0 and line[0] == '#':
    print('Es un comentario')
````

## Glosario

- **contador (counter)**: Una variable usada para contar algo, usualmente inicializada a cero y luego incrementada.

- **cadena vacía (empty string)**: Una cadena sin caracteres y de longitud 0, representada por dos comillas.

- **bandera (flag)**: Una variable booleana usada para indicar si una condición es verdadera o falsa.

- **invocación (invocation)**: Una sentencia que llama a un método.

- **inmutable (immutable)**: La propiedad de una secuencia cuyos ítems no pueden ser asignados.

- **índice (index)**: Un valor entero usado para seleccionar un ítem en una secuencia, como un carácter en una cadena.

- **ítem (item)**: Uno de los valores de una secuencia.

- **método (method)**: Una función que está asociada con un objeto y se llama usando la notación de punto.

- **objeto (object)**: Algo a lo que una variable puede referirse. Por ahora, puedes usar "objeto" y "valor" indistintamente.

- **búsqueda (search)**: Un patrón de recorrido que se detiene cuando encuentra lo que está buscando.

- **secuencia (sequence)**: Un conjunto ordenado; es decir, un conjunto de valores donde cada valor se identifica mediante un índice entero.

- **rebanada (slice)**: Una parte de una cadena especificada por un rango de índices.

- **recorrer (traverse)**: Iterar a través de los ítems de una secuencia, realizando una operación similar sobre cada uno.
