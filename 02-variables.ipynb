## Valores y tipos

Un *valor* es una de las unidades básicas con las que trabaja un programa, como una letra, un número o una cadena de texto. Los valores que hemos visto hasta ahora son 1, 2 y "¡Hola, mundo!".

Estos valores pertenecen a diferentes *tipos*: 2 es un entero, mientras que "¡Hola, mundo!" es una *cadena* o `string`, llamada así porque contiene una secuencia de letras. El intérprete y el programador pueden identificar las cadenas porque están encerradas entre comillas.

La instrucción `print` también funciona con enteros. Usamos el comando `python` para iniciar el intérprete.

```python
python
>>> print(4)
4
```

Si no estamos seguros del tipo de un valor, el intérprete puede indicárnoslo.

```python
>>> type('Hola, mundo!')
<class 'str'>
>>> type(17)
<class 'int'>
```

No debería sorprendernos que las cadenas pertenezcan al tipo `str` y los enteros al tipo `int`. Menos evidente es el caso de los números con punto decimal, que pertenecen a un tipo llamado `float`, porque se representan usando el formato de *punto flotante*.

```python
>>> type(3.2)
<class 'float'>
```

¿Qué ocurre con valores como "17" y "3.2"? Aunque parecen números, están rodeados por comillas, por lo que Python los reconoce como cadenas de texto.

```python
>>> type('17')
<class 'str'>
>>> type('3.2')
<class 'str'>
```

Son cadenas.

Cuando escribimos un entero grande, puede resultar tentador usar comas entre grupos de tres cifras, como en `1,000,000`. En Python esto no es un entero válido; sin embargo, la expresión es legal y produce una secuencia de números separados por comas.

```python
>>> print(1,000,000)
1 0 0
```

Este comportamiento ilustra un caso importante: Python ejecuta la instrucción sin mostrar un error, pero el resultado no corresponde a la intención original del programador. Este es un ejemplo de *error semántico*.


````python
print(4)
print(type('Hola, mundo!'))
print(type(17))
print(type(3.2))
print(type('17'))
print(type('3.2'))
print(1,000,000)
````

## Variables

Una de las características más potentes de un lenguaje de programación es la capacidad de manipular *variables*. Una variable es un nombre que hace referencia a un valor concreto y puede cambiar durante la ejecución del programa.

Una *instrucción de asignación* crea nuevas variables y les asigna valores:

```python
>>> mensaje = 'Y ahora, algo completamente diferente'
>>> n = 17
>>> pi = 3.1415926535897931
```

Este ejemplo realiza tres asignaciones. La primera asigna una cadena a la variable `mensaje`; la segunda asigna el entero 17 a `n`; la tercera asigna el valor aproximado de $\pi$ a `pi`.

Para mostrar el contenido de una variable, puede usarse una instrucción `print`:

```python
>>> print(n)
17
>>> print(pi)
3.141592653589793
```

El tipo de una variable coincide con el tipo del valor al que hace referencia.

```python
>>> type(mensaje)
<class 'str'>
>>> type(n)
<class 'int'>
>>> type(pi)
<class 'float'>
```

### Reflexión

La asignación no solo almacena información; también define la relación entre nombre y valor. En programación, esta relación es fundamental para representar datos, construir cálculos y mantener estados dentro de un programa.

````python
mensaje = 'Y ahora, algo completamente diferente'
n = 17
pi = 3.1415926535897931
print(n)
print(pi)
print(type(mensaje))
print(type(n))
print(type(pi))
````

## Nombres de variables y palabras clave

Los programadores normalmente eligen nombres para sus variables que sean significativos y documenten para qué se usará la variable.

Los nombres de variables pueden ser arbitrariamente largos. Pueden contener letras y números, pero no pueden empezar con un número. Es legal usar letras mayúsculas, pero es una buena idea comenzar los nombres de variables con minúscula (más adelante verás por qué).

El carácter de subrayado (_) puede aparecer en un nombre. A menudo se usa en nombres con varias palabras, como `mi_nombre` o `velocidad_del_swallow_no_cargado`. Los nombres de variables pueden comenzar con un guion bajo, pero generalmente evitamos hacerlo salvo que estemos escribiendo código de biblioteca para otros.

Si le das a una variable un nombre ilegal, obtienes un error de sintaxis:

```python
>>> 76trombones = 'gran desfile'
SyntaxError: invalid syntax
>>> more@ = 1000000
SyntaxError: invalid syntax
>>> class = 'Zymurgia Teórica Avanzada'
SyntaxError: invalid syntax
```

`76trombones` es ilegal porque comienza con un número. `more@` es ilegal porque contiene un carácter no permitido, @. Pero, ¿qué tiene de malo `class`?

Resulta que `class` es una de las *palabras clave* de Python. El intérprete usa las palabras clave para reconocer la estructura del programa y no pueden usarse como nombres de variables.

Python reserva 35 palabras clave:

```text
False      await      else       import     pass
None       break      except     in         raise
True       class      finally    is         return
and        continue   for        lambda     try
as         def        from       nonlocal   while
assert     del        global     not        with
async      elif       if         or         yield
```

## Sentencias

Una *sentencia* es una unidad de código que el intérprete de Python puede ejecutar. En esta etapa del curso, hemos visto dos tipos fundamentales: la instrucción `print` y la asignación de variables.

Cuando se escribe una sentencia en modo interactivo, el intérprete la ejecuta y muestra el resultado, si existe uno. En modo script, un archivo puede contener varias sentencias en secuencia. Cada una se ejecuta en orden, y el efecto observable puede depender del flujo del programa.

Por ejemplo, el siguiente script:

```python
print(1)
x = 2
print(x)
```

produce la salida:

```text
1
2
```

La instrucción de asignación no produce salida visible por sí misma; su función es almacenar un valor en una variable para su uso posterior.

### Diferencia entre expresión y sentencia

Una *expresión* produce un valor. Una *sentencia* realiza una acción. Por ejemplo, `x + 1` es una expresión; `x = x + 1` es una sentencia de asignación. Esta distinción es central para comprender la semántica de Python.

````python
print(1)
x = 2
print(x)
````

## Operadores y operandos

Los *operadores* son símbolos especiales que representan cálculos, como suma, resta, multiplicación o división. Los valores sobre los que actúa un operador se denominan *operandos*.

Los operadores `+`, `-`, `*`, `/` y `**` realizan suma, resta, multiplicación, división y exponenciación, respectivamente, como se muestra a continuación:

```python
20 + 32
hora - 1
hora * 60 + minuto
minuto / 60
5 ** 2
(5 + 9) * (15 - 7)
```

Existe una diferencia clave entre Python 2 y Python 3 en el operador de división. En Python 3, el cociente de una división entre enteros produce un valor de punto flotante:

```python
>>> minuto = 59
>>> minuto / 60
0.9833333333333333
```

En Python 2, la división entre enteros truncaba el resultado a entero:

```python
>>> minuto = 59
>>> minuto / 60
0
```

Para obtener el mismo comportamiento esperado en Python 3, se usa la división entera con `//`:

```python
>>> minuto = 59
>>> minuto // 60
0
```

En Python 3, la división entera se comporta de manera más intuitiva para cálculos algebraicos y anotaciones de programación, y es comparable a la operación de división que se esperaría en una calculadora.

### Importancia en modelado analítico

En contexto de análisis estadístico o científico, una comprensión correcta de los operadores aritméticos permite representar con precisión fórmulas, tasas, proporciones y transformaciones de variables.

````python
hora = 3
minuto = 59
print(20 + 32)
print(hora - 1)
print(hora * 60 + minuto)
print(minuto / 60)
print(5 ** 2)
print((5 + 9) * (15 - 7))
print(minuto // 60)
````

## Expresiones

Una *expresión* es una combinación de valores, variables y operadores que produce un resultado. A la práctica, cualquier literal, variable o combinación de ambos puede considerarse una expresión válida. Por ejemplo, si la variable `x` ya tiene un valor asignado:

```python
17
x
x + 17
```

Si se escribe una expresión en modo interactivo, el intérprete la *evalúa* y muestra el resultado:

```python
>>> 1 + 1
2
```

En un script, una expresión aislada no genera efecto visible; no se imprime ni se asigna. Este comportamiento suele resultar confuso para principiantes, pero es una característica fundamental del diseño de Python.

## Orden de operaciones

Cuando aparece más de un operador en una expresión, el orden de evaluación depende de las *reglas de precedencia*. Para operadores matemáticos, Python sigue la convención general de la aritmética.

El acrónimo *PEMDAS* es una forma útil de recordar estas reglas:

- *P*aréntesis tienen la prioridad más alta y permiten controlar explícitamente el orden de evaluación. Por eso, `2 * (3 - 1)` es 4 y `(1 + 1) ** (5 - 2)` es 8.
- *E*xponenciación tiene la siguiente prioridad, de modo que `2 ** 1 + 1` es 3, no 4, y `3 * 1 ** 3` es 3, no 27.
- *M*ultiplicación y *D*ivisión tienen la misma prioridad, y ambas son superiores a *A*ddición y *S*ustracción.
- Cuando dos operadores tienen la misma precedencia, se evalúan de izquierda a derecha. Así, `5 - 3 - 1` se interpreta como `(5 - 3) - 1`, no como `5 - (3 - 1)`.

Cuando existan dudas, es recomendable usar paréntesis para hacer explícito el orden de cálculo y evitar errores semánticos.

## Operador módulo

El *operador módulo* trabaja con enteros y devuelve el resto que resulta al dividir el primer operando por el segundo. En Python, el operador módulo se representa con `%`.

```python
>>> cociente = 7 // 3
>>> print(cociente)
2
>>> resto = 7 % 3
>>> print(resto)
1
```

Por tanto, 7 dividido por 3 es 2, con un residuo de 1.

El operador módulo es muy útil en múltiples contextos. Por ejemplo, permite verificar si un número es divisible por otro: si `x % y` es cero, entonces `x` es divisible por `y`.

También permite extraer el último dígito o los últimos dígitos de un número. En particular, `x % 10` devuelve la última cifra de `x` en base 10, mientras que `x % 100` devuelve los dos últimos dígitos.

````python
cociente = 7 // 3
resto = 7 % 3
print(cociente)
print(resto)
print(12345 % 10)
print(12345 % 100)
````

## Operaciones con cadenas

El operador `+` funciona con cadenas, pero no representa suma en el sentido matemático. En este contexto realiza *concatenación*: une dos cadenas de texto de extremo a extremo.

```python
>>> primero = 10
>>> segundo = 15
>>> print(primero + segundo)
25
>>> primero = '100'
>>> segundo = '150'
>>> print(primero + segundo)
100150
```

El operador `*` también puede aplicarse a cadenas, multiplicando la cadena por un número entero.

```python
>>> primero = 'Prueba '
>>> segundo = 3
>>> print(primero * segundo)
Prueba Prueba Prueba
```

Este comportamiento es útil para generar textos repetidos, etiquetas o patrones de salida en scripts y aplicaciones interactivas.

````python
primero = 10
segundo = 15
print(primero + segundo)
primero = '100'
segundo = '150'
print(primero + segundo)
primero = 'Prueba '
segundo = 3
print(primero * segundo)
````

## Solicitud de entrada del usuario

En muchos programas es necesario obtener información directamente del usuario. Python incluye la función incorporada `input`, que captura texto desde el teclado. Cuando se invoca, el programa se pausa y espera a que el usuario escriba una respuesta. Al presionar `Enter`, la función devuelve lo ingresado como una cadena de texto.

```python
>>> entrada = input()
Algo tonto
>>> print(entrada)
Algo tonto
```

Antes de solicitar una respuesta, es recomendable mostrar un mensaje guía. Esto se hace pasando una cadena como argumento a `input`:

```python
>>> nombre = input('¿Cómo te llamas?\n')
¿Cómo te llamas?
Chuck
>>> print(nombre)
Chuck
```

La secuencia `\n` al final del mensaje representa un salto de línea, que hace que la entrada del usuario aparezca en la siguiente línea.

Si se espera que el usuario ingrese un número entero, puede convertir la cadena resultante con la función `int()`:

```python
>>> prompt = '¿Cuál... es la velocidad del ala de una golondrina sin carga?\n'
>>> velocidad = input(prompt)
¿Cuál... es la velocidad del ala de una golondrina sin carga?
17
>>> int(velocidad)
17
>>> int(velocidad) + 5
22
```

Si el usuario escribe un valor no numérico, Python lanzará un error de tipo `ValueError`.

```python
>>> velocidad = input(prompt)
¿Cuál... es la velocidad del ala de una golondrina sin carga?
¿Qué quieres decir, una golondrina africana o europea?
>>> int(velocidad)
ValueError: invalid literal for int() with base 10: '¿Qué quieres decir, una golondrina africana o europea?'
```

Este tipo de situaciones es habitual en la programación práctica y será abordado más adelante con técnicas de validación y manejo de errores.

````python
# Ejemplo de entrada del usuario
nombre = input('¿Cómo te llamas?\n')
print(nombre)

mensaje = '¿Cuál... es la velocidad del ala de una golondrina sin carga?\n'
velocidad = int(input(mensaje))
print(velocidad + 5)
````

## Comentarios

A medida que los programas crecen y se vuelven más complejos, se vuelven más difíciles de leer. Los lenguajes formales son densos y, con frecuencia, resulta difícil interpretar un bloque de código sin un contexto adicional.

Por esta razón, es una buena práctica incluir notas explicativas en el código. En Python, estas notas se denominan *comentarios* y comienzan con el símbolo `#`:

```python
# calcula el porcentaje de la hora transcurrida
porcentaje = (minuto * 100) / 60
```

En este caso, el comentario aparece en una línea independiente. También puede escribirse al final de una línea de código:

```python
porcentaje = (minuto * 100) / 60     # porcentaje de una hora
```

Todo lo que está entre `#` y el final de la línea es ignorado por el intérprete y no afecta la ejecución del programa.

Los comentarios son más útiles cuando documentan decisiones no obvias, supuestos o relaciones difíciles de inferir a simple vista. El lector suele deducir qué hace el código; lo más valioso es explicar por qué se ha escrito así.

Ejemplos:

```python
v = 5     # asigna 5 a v
```

Este comentario es redundante con el código y aporta poca información.

```python
v = 5     # velocidad en metros por segundo
```

Este comentario sí aporta contexto útil que no se observa directamente en la instrucción.

### Buenas prácticas

Los nombres de variables claros ayudan a reducir la necesidad de comentarios, pero los nombres demasiado largos pueden hacer que expresiones complejas sean difíciles de leer. El equilibrio entre claridad y concisión es una habilidad esencial en programación.

## Selección de nombres de variables mnemotécnicos

Mientras se respeten las reglas básicas de nomenclatura y se eviten las palabras reservadas, el programador dispone de gran libertad para escoger nombres. Sin embargo, esta libertad puede generar ambigüedad si los nombres no expresan bien la intención del código.

Por ejemplo, los siguientes tres programas cumplen la misma función, pero presentan diferencias importantes en legibilidad y comprensión:

```python
a = 35.0
b = 12.50
c = a * b
print(c)
```

```python
horas = 35.0
tarifa = 12.50
pago = horas * tarifa
print(pago)
```

```python
x1q3z9ahd = 35.0
x1q3z9afd = 12.50
x1q3p9afd = x1q3z9ahd * x1q3z9afd
print(x1q3p9afd)
```

Aunque Python interpreta los tres programas de la misma forma, un humano entiende mucho más rápido la intención del segundo bloque. Eso ocurre porque los nombres elegidos reflejan el significado de los datos almacenados.

Llamamos a estos nombres *mnemotécnicos*. El término *mnemotécnico* hace referencia a una ayuda para la memoria. Elegimos nombres con intención para recordar por qué se creó la variable y qué función cumple en el programa.

No obstante, el uso de nombres demasiado descriptivos puede dificultar la lectura inicial de un código a principiantes, porque algunos tokens pueden confundirse con partes del lenguaje. Por ejemplo:

```python
for palabra in palabras:
    print(palabra)
```

Aquí puede resultar útil distinguir entre palabras reservadas (`for`, `in`) y variables elegidas por el programador (`palabra`, `palabras`).

El siguiente ejemplo es equivalente desde el punto de vista sintáctico, pero más difícil de interpretar para un lector inicial:

```python
for rebanada in pizza:
    print(rebanada)
```

A largo plazo, el lector aprende a reconocer rápidamente las palabras reservadas y a separar claramente qué parte es sintaxis y qué parte es contenido definido por el programador.

### Principio didáctico

Un buen nombre de variable debe ser:

- claro;
- breve;
- consistente;
- expresivo del dominio del problema.

Este criterio es especialmente relevante en investigación cuantitativa, análisis de datos y modelado estadístico.

## Depuración

En este punto, el error de sintaxis más frecuente es un nombre de variable no válido, como `class` o `yield`, o caracteres ilegales como `odd~job` y `US$`.

Si se incluye un espacio dentro de un nombre de variable, Python lo interpreta como dos operandos sin operador:

```python
>>> bad name = 5
SyntaxError: invalid syntax
```

Los mensajes de error de sintaxis suelen ser poco descriptivos. El más común es: `SyntaxError: invalid syntax`.

El error de ejecución más habitual es el intento de usar una variable antes de asignarle un valor. Esto puede ocurrir si se escribe mal el nombre de la variable:

```python
>>> principal = 327.68
>>> interest = principle * rate
NameError: name 'principle' is not defined
```

Los nombres de variables son sensibles a mayúsculas y minúsculas. Por ejemplo, `LaTeX` no es equivalente a `latex`.

Otro tipo de error frecuente es el semántico: la expresión se ejecuta, pero el resultado no corresponde a la intención del programador. Un caso clásico es el orden de las operaciones. Por ejemplo, para evaluar $1/2\pi$, podrías escribir:

```python
>>> 1.0 / 2.0 * pi
```

Pero, como Python realiza primero la división, el cálculo equivale a $\pi / 2$, no a $1 / (2\pi)$. El intérprete no puede inferir la intención del autor y, por ello, no emite un error; simplemente entrega un resultado incorrecto.

### Recomendación

Cuando una expresión resulta ambigua, conviene usar paréntesis para hacer explícito el orden de cálculo.

## Glosario

- **asignación**: instrucción que asigna un valor a una variable.
- **concatenar**: unir dos operandos de extremo a extremo.
- **comentario**: información escrita en un programa para otros programadores o lectores del código, y que no tiene efecto sobre la ejecución.
- **evaluar**: simplificar una expresión realizando las operaciones en el orden correcto para producir un solo valor.
- **expresión**: combinación de valores, variables y operadores que produce un resultado.
- **punto flotante**: tipo que representa números con parte fraccionaria.
- **entero**: tipo que representa números sin parte decimal.
- **palabra clave**: palabra reservada utilizada por el compilador o intérprete para analizar la estructura del programa.
- **mnemotécnico**: nombre de variable diseñado para recordar el propósito o significado del dato almacenado.
- **operador módulo**: operador `%`, que devuelve el residuo de la división entera.
- **operando**: valor sobre el cual actúa un operador.
- **operador**: símbolo especial que representa una operación simple, como suma, multiplicación o concatenación.
- **reglas de precedencia**: conjunto de reglas que determina el orden de evaluación de una expresión con varios operadores.
- **sentencia**: unidad de código que representa una acción ejecutable.
- **cadena**: tipo que representa secuencias de caracteres.
- **tipo**: categoría de valores. En Python, los tipos básicos incluyen `int`, `float` y `str`.
- **valor**: unidad básica de datos que un programa manipula.
- **variable**: nombre que referencia a un valor.
