<h2>Leer los datos</h2>
<p>
Usamos la función <code>pandas.read_csv()</code> para leer archivos CSV. Sin embargo, en esta versión del laboratorio, que funciona en JupyterLite, el conjunto de datos debe descargarse a la interfaz con el código que se muestra a continuación.
</p>

Las funciones siguientes descargan el conjunto de datos en tu navegador:

````python
# Descarga un archivo desde una URL y lo guarda en disco.
# Funciona igual en JupyterLite (kernel del navegador) y en un kernel local
# (Anaconda, VS Code, etc.). Mantiene la firma asincrona, de modo que la celda
# que hace "await download(file_path, 'auto.csv')" sigue funcionando igual.

try:
    # Kernel del navegador: pyodide solo existe dentro de JupyterLite.
    from pyodide.http import pyfetch

    async def download(url, filename):
        response = await pyfetch(url)
        if response.status == 200:
            with open(filename, "wb") as f:
                f.write(await response.bytes())
        else:
            raise Exception(f"Download failed with status {response.status}")

except ModuleNotFoundError:
    # Kernel local (CPython): usamos la libreria estandar, sin pyodide.
    import asyncio
    from urllib.request import Request, urlopen

    def _descargar(url, filename):
        peticion = Request(url, headers={"User-Agent": "Mozilla/5.0"})
        with urlopen(peticion) as respuesta, open(filename, "wb") as f:
            f.write(respuesta.read())

    async def download(url, filename):
        # en un hilo aparte para no bloquear el bucle de eventos del cuaderno
        await asyncio.to_thread(_descargar, url, filename)
````

````python
file_path='https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-SkillsNetwork/labs/Data%20files/auto.csv'
````

Para obtener el conjunto de datos, usa la función download() definida arriba:

<h3>Agregar encabezados</h3>
<p>
Observa el conjunto de datos. Pandas asignó automáticamente los encabezados con números enteros que empiezan en 0.
</p>
<p>
Para describir mejor los datos puedes asignar encabezados propios. Esta información está disponible en:  <a href="https://archive.ics.uci.edu/ml/datasets/Automobile" target="_blank">https://archive.ics.uci.edu/ml/datasets/Automobile</a>.
</p>
<p>
Por lo tanto, debes agregar los encabezados manualmente.
</p>
<p>
Primero, crea una lista "headers" que incluya todos los nombres de las columnas en orden.
Luego usa <code>dataframe.columns = headers</code> para reemplazar los encabezados por la lista que creaste.
</p>

````python
# crear la lista de encabezados
headers = ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]
print("headers\n", headers)
````

 Reemplaza los encabezados y vuelve a revisar el dataframe:

````python
df.columns = headers
df.columns
````

También puedes ver las primeras 10 filas del dataframe actualizado y notar que los encabezados ya cambiaron.

````python
df.head(10)
````

Ahora debemos reemplazar el símbolo "?" por NaN, para que dropna() pueda eliminar los valores faltantes:

````python
df1 = df.replace('?', np.nan)
````

Puedes eliminar los valores faltantes de la columna "price" así:

````python
df = df1.dropna(subset=["price"]).reset_index(drop=True)
df.head(20)
````

Aquí, `axis=0` significa que se eliminará toda la fila donde el campo 'price' sea NaN.

Ahora has leído correctamente el conjunto de datos original y agregado los encabezados correctos al dataframe.

<h2>Tipos de datos</h2>
<p>
Los datos tienen diversos tipos.<br>

Los tipos principales que se almacenan en los dataframe de Pandas son <b>object</b>, <b>float</b>, <b>int</b>, <b>bool</b> y <b>datetime64</b>. Para conocer mejor cada atributo, siempre debes saber cuál es el tipo de dato de cada columna. En Pandas:
</p>

````python
df.dtypes
````

Devuelve una serie con el tipo de dato de cada columna.

````python
# revisar el tipo de dato del dataframe "df" con .dtypes
print(df.dtypes)
````

<p>
Como se muestra arriba, puedes ver claramente que el tipo de dato de "symboling" y "curb-weight" es <code>int64</code>, el de "normalized-losses" es <code>object</code> y el de "wheel-base" es <code>float64</code>, etc.
</p>
<p>
Estos tipos de datos se pueden cambiar; aprenderás cómo hacerlo en un módulo posterior.
</p>

<h2>Describe</h2>
Si queremos obtener un resumen estadístico de cada columna, como el conteo, la media, la desviación estándar, etc., usamos el método describe:

Este método entrega varias estadísticas de resumen y excluye los valores <code>NaN</code> (Not a Number, es decir, "no es un número").

````python
df.describe()
````

<p>
Esto muestra el resumen estadístico de todas las columnas numéricas (int, float).<br>

Por ejemplo, el atributo "symboling" tiene 205 datos, la media de esta columna es 0.83, la desviación estándar es 1.25, el valor mínimo es -2, el percentil 25 es 0, el percentil 50 es 1, el percentil 75 es 2 y el valor máximo es 3.
<br>

Sin embargo, ¿y si también quieres revisar todas las columnas, incluidas las de tipo object?
<br><br>

Puedes agregar el argumento <code>include = "all"</code> dentro del paréntesis. Inténtalo de nuevo.

</p>

````python
# describir todas las columnas de "df" 
df.describe(include = "all")
````

<p>
Ahora entrega el resumen estadístico de todas las columnas, incluidas las de tipo object.<br>

Ahora puedes ver cuántos valores únicos hay, cuál es el valor más frecuente (top) y cuál es la frecuencia de ese valor en las columnas de tipo object.<br>

Algunos valores de la tabla anterior muestran "NaN". Esos números no están disponibles para cierto tipo de columna.<br>
</p>
