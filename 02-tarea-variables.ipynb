# Tarea 1: Variables, expresiones y sentencias

## Objetivo

Aplicar los conceptos básicos de Python para escribir programas simples que manipulen variables, realicen cálculos aritméticos y capturen información del usuario.

## Instrucciones

1. Escribe un programa en Python que solicite el nombre de una persona y muestre un mensaje de bienvenida.
2. Escribe un programa que solicite la cantidad de horas trabajadas y la tarifa por hora, y luego calcule el salario bruto.
3. Escribe un programa que solicite una temperatura en grados Celsius y la convierta a Fahrenheit.
4. Crea un ejemplo con variables y operadores aritméticos que muestre el cálculo del área de un rectángulo.
5. Explica, en 4–6 líneas, la diferencia entre una variable, una expresión y una sentencia en Python.

## Requisitos formales

- Usa nombres de variables claros y significativos.
- Incluye comentarios breves cuando sea necesario.
- Verifica que los resultados sean correctos ejecutando el código.
- Entrega el archivo en formato `.py` o como notebook de Jupyter.

## Ejemplo de salida esperada

```text
Ingrese su nombre: Ana
Bienvenida, Ana
```

```text
Ingrese horas trabajadas: 40
Ingrese tarifa por hora: 25
Salario bruto: 1000.0
```

```text
Ingrese temperatura en °C: 20
Temperatura en °F: 68.0
```
