# Gráficos de líneas (Series/Dataframe) <a id="4"></a>


**¿Qué es un gráfico de líneas y para qué sirve?**

Un gráfico de líneas o gráfico lineal es un tipo de gráfico que muestra información como una serie de puntos de datos llamados "marcadores" conectados por segmentos de línea recta. Es un tipo de gráfico básico y común en muchos campos.
Usa un gráfico de líneas cuando tengas un conjunto de datos continuo. Estos gráficos son los más adecuados para visualizaciones de tendencias de datos a lo largo del tiempo.


**Empecemos con un caso de estudio:**

En 2010, Haití sufrió un terremoto catastrófico de magnitud 7.0. El sismo causó una devastación generalizada y pérdida de vidas: cerca de tres millones de personas se vieron afectadas por este desastre natural. Como parte del esfuerzo humanitario de Canadá, el Gobierno canadiense redobló sus esfuerzos para aceptar refugiados de Haití. Podemos visualizar rápidamente este esfuerzo con un gráfico de `líneas`:

**Pregunta:** Grafica una línea de la inmigración desde Haití usando `df.plot()`.


Primero extraeremos la serie de datos de Haití.


````python
#Como convertimos los años a texto,
#declaremos una variable que nos permita invocar fácilmente todo el rango de años:
years = list(map(str, range(1980, 2014)))
#creando la serie de datos
haiti = df_can.loc['Haiti', years] # pasando los años 1980 - 2013 para excluir la columna 'total'
haiti.head()
````

A continuación, graficaremos una línea añadiendo `.plot()` al dataframe `haiti`.


````python
haiti.plot()
````

*pandas* llenó automáticamente el eje x con los valores del índice (los años) y el eje y con los valores de la columna (la población). <br>

Además, etiquetemos los ejes x e y usando `plt.title()`, `plt.ylabel()` y `plt.xlabel()` de la siguiente manera:


````python
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of immigrants')
plt.xlabel('Years')

plt.show() # se necesita esta línea para mostrar los cambios hechos a la figura
````

Podemos notar claramente cómo el número de inmigrantes desde Haití se disparó a partir de 2010, cuando Canadá redobló sus esfuerzos para aceptar refugiados de Haití. Anotemos este pico en el gráfico usando el método `plt.text()`.<br>
<br>
Sin embargo, observa que los años son de tipo *string* (texto). Primero cambiemos el tipo de los valores del índice a *integer* (entero).


````python
haiti.index = haiti.index.map(int) 
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

# anotar el terremoto de 2010. 
# sintaxis: plt.text(x, y, label)
plt.text(2000, 6000, '2010 Earthquake') # ver la nota más abajo

plt.show() 
````

¡Con apenas unas líneas de código pudiste identificar y visualizar rápidamente el pico de inmigración!

Nota rápida sobre los valores x e y en `plt.text(x, y, label)`:
    
     Como el eje x (los años) es de tipo 'integer', indicamos x como un año. El eje y (número de inmigrantes) es de tipo 'integer', así que simplemente podemos indicar el valor y = 6000.
    
```python
    plt.text(2000, 6000, '2010 Earthquake') # años almacenados como tipo int
```
    Si los años estuvieran almacenados como tipo 'string', tendríamos que indicar x como la posición del índice del año. Por ejemplo, el índice 20 corresponde al año 2000, ya que es el año número 20 con año base 1980.
```python
    plt.text(20, 6000, '2010 Earthquake') # años almacenados como tipo int
```
    Cubriremos métodos de anotación avanzados en módulos posteriores.

### Otros gráficos

¡Felicidades! Has aprendido a procesar datos con Python y a crear un gráfico de líneas con Matplotlib. Existen muchos otros estilos de gráficos además del gráfico de líneas predeterminado, y todos se pueden invocar pasando la palabra clave `kind` a `plot()`. La lista completa de gráficos disponibles es la siguiente:

* `bar` para gráficos de barras verticales
* `barh` para gráficos de barras horizontales
* `hist` para histogramas
* `box` para diagramas de caja (boxplot)
* `kde` o `density` para gráficos de densidad
* `area` para gráficos de área
* `pie` para gráficos de sectores (pie)
* `scatter` para gráficos de dispersión
* `hexbin` para gráficos hexbin
