## Abrir archivos

Cuando queremos leer o escribir un archivo (digamos, en tu disco duro), primero debemos *abrir* el archivo. Abrir el archivo se comunica con tu sistema operativo, que sabe dónde están almacenados los datos de cada archivo. Cuando abres un archivo, le estás pidiendo al sistema operativo que encuentre el archivo por su nombre y se asegure de que el archivo existe. En este ejemplo, abrimos el archivo *mbox.txt*, que debería estar almacenado en la misma carpeta en la que estás cuando inicias Python. Puedes descargar este archivo desde [www.py4e.com/code3/mbox.txt](http://www.py4e.com/code3/mbox.txt)

*(En este notebook el archivo `mbox.txt` ya está copiado junto al notebook.)*


````python
fhand = open('mbox.txt')
print(fhand)   # <_io.TextIOWrapper name='mbox.txt' mode='r' encoding='cp1252'>
````

Si el `open` tiene éxito, el sistema operativo nos devuelve un *manejador de archivo* (file handle). El manejador de archivo no son los datos reales contenidos en el archivo, sino un "manejador" que podemos usar para leer los datos. Se te da un manejador si el archivo solicitado existe y tienes los permisos adecuados para leerlo.


![Un manejador de archivo](../../images/handle.svg)


Si el archivo no existe, `open` fallará con un traceback y no obtendrás un manejador para acceder al contenido del archivo:

```
>>> fhand = open('stuff.txt')
Traceback (most recent call last):
File "<stdin>", line 1, in <module>
FileNotFoundError: [Errno 2] No such file or directory: 'stuff.txt'
```

Más adelante usaremos `try` y `except` para manejar de forma más elegante la situación en la que intentamos abrir un archivo que no existe.


## Archivos de texto y líneas

Un archivo de texto puede pensarse como una secuencia de líneas, de la misma manera que una cadena de Python puede pensarse como una secuencia de caracteres. Por ejemplo, esta es una muestra de un archivo de texto que registra la actividad de correo de varias personas en un equipo de desarrollo de un proyecto de código abierto:

```
From stephen.marquard@uct.ac.za Sat Jan  5 09:14:16 2008
Return-Path: <postmaster@collab.sakaiproject.org>
Date: Sat, 5 Jan 2008 09:12:18 -0500
To: source@collab.sakaiproject.org
From: stephen.marquard@uct.ac.za
Subject: [sakai] svn commit: r39772 - content/branches/
Details: http://source.sakaiproject.org/viewsvn/?view=rev&rev=39772
...
```

El archivo completo de las interacciones de correo está disponible en

[www.py4e.com/code3/mbox.txt](http://www.py4e.com/code3/mbox.txt)

y una versión acortada del archivo está disponible en

[www.py4e.com/code3/mbox-short.txt](http://www.py4e.com/code3/mbox-short.txt)

Estos archivos están en un formato estándar para un archivo que contiene múltiples mensajes de correo. Las líneas que comienzan con "From " separan los mensajes, y las líneas que comienzan con "From:" son parte de los mensajes. Para más información sobre el formato mbox, consulta <https://en.wikipedia.org/wiki/Mbox>.

Para dividir el archivo en líneas, hay un carácter especial que representa el "final de la línea", llamado el carácter de *nueva línea* (newline).

En Python, representamos el carácter de *nueva línea* como una barra invertida-n en las constantes de cadena. Aunque parezcan dos caracteres, en realidad es un solo carácter. Cuando miramos la variable escribiendo "stuff" en el intérprete, nos muestra el `\n` en la cadena, pero cuando usamos `print` para mostrar la cadena, vemos la cadena dividida en dos líneas por el carácter de nueva línea.


````python
stuff = 'Hello\nWorld!'
stuff            # 'Hello\nWorld!'  (así lo muestra el intérprete)
print(stuff)     # Hello
                 # World!
stuff = 'X\nY'
print(stuff)     # X
                 # Y
len(stuff)       # 3
````

También puedes ver que la longitud de la cadena `X\nY` es *tres* caracteres, porque el carácter de nueva línea es un solo carácter.

Así que cuando miramos las líneas de un archivo, necesitamos *imaginar* que hay un carácter especial invisible llamado nueva línea al final de cada línea que marca el final de la línea.

Así que el carácter de nueva línea separa los caracteres del archivo en líneas.


## Leer archivos

Aunque el *manejador de archivo* no contiene los datos del archivo, es bastante fácil construir un bucle `for` para leer y contar cada una de las líneas de un archivo (`open.py`):


````python
fhand = open('mbox-short.txt')
count = 0
for line in fhand:
    count = count + 1
print('Line Count:', count)
````

*(El programa lee el archivo `mbox-short.txt`, que está junto a este notebook.)*

Podemos usar el manejador de archivo como la secuencia en nuestro bucle `for`. Nuestro bucle `for` simplemente cuenta el número de líneas del archivo y lo imprime. La traducción aproximada del bucle `for` al lenguaje natural es: "para cada línea del archivo representado por el manejador de archivo, suma uno a la variable `count`".

La razón por la que la función `open` no lee el archivo completo es que el archivo podría ser bastante grande, con muchos gigabytes de datos. La sentencia `open` tarda lo mismo independientemente del tamaño del archivo. El bucle `for` es el que realmente hace que los datos se lean del archivo.

Cuando el archivo se lee usando un bucle `for` de esta manera, Python se encarga de dividir los datos del archivo en líneas separadas usando el carácter de nueva línea. Python lee cada línea hasta el salto de línea e incluye el salto de línea como último carácter en la variable `line` en cada iteración del bucle `for`.

Debido a que el bucle `for` lee los datos de una línea cada vez, puede leer y contar eficientemente las líneas de archivos muy grandes sin quedarse sin memoria principal para almacenar los datos. El programa anterior puede contar las líneas de un archivo de cualquier tamaño usando muy poca memoria, ya que cada línea se lee, se cuenta y luego se descarta.

Si sabes que el archivo es relativamente pequeño en comparación con el tamaño de tu memoria principal, puedes leer el archivo completo en una sola cadena usando el método `read` sobre el manejador de archivo.


````python
fhand = open('mbox-short.txt')
inp = fhand.read()
print(len(inp))     # 94626
print(inp[:20])     # From stephen.marquar
````

En este ejemplo, todo el contenido (los 94.626 caracteres) del archivo *mbox-short.txt* se lee directamente en la variable `inp`. Usamos rebanado de cadenas para imprimir los primeros 20 caracteres de los datos de la cadena almacenados en `inp`.

Cuando el archivo se lee de esta manera, todos los caracteres, incluyendo todas las líneas y caracteres de nueva línea, forman una gran cadena en la variable `inp`. Es una buena idea almacenar la salida de `read` en una variable porque cada llamada a `read` agota el recurso:


````python
fhand = open('mbox-short.txt')
print(len(fhand.read()))   # 94626
print(len(fhand.read()))   # 0
````

Recuerda que esta forma de la función `open` solo debería usarse si los datos del archivo caben cómodamente en la memoria principal de tu computador. Si el archivo es demasiado grande para caber en la memoria principal, deberías escribir tu programa para leer el archivo en trozos usando un bucle `for` o `while`.


## Buscar dentro de un archivo

Cuando buscas dentro de los datos de un archivo, un patrón muy común es leer el archivo, ignorando la mayoría de las líneas y procesando solo las líneas que cumplen una condición particular. Podemos combinar el patrón de lectura de un archivo con métodos de cadenas para construir mecanismos de búsqueda sencillos.

Por ejemplo, si quisiéramos leer un archivo e imprimir solo las líneas que comienzan con el prefijo "From:", podríamos usar el método de cadenas *startswith* para seleccionar solo aquellas líneas con el prefijo deseado (`search1.py`):


````python
fhand = open('mbox-short.txt')
for line in fhand:
    if line.startswith('From:'):
        print(line)
````

Cuando este programa se ejecuta, obtenemos la siguiente salida:

```
From: stephen.marquard@uct.ac.za

From: louis@media.berkeley.edu

From: zqian@umich.edu

From: rjlowe@iupui.edu
...
```

La salida se ve muy bien, ya que las únicas líneas que estamos viendo son las que comienzan con "From:", pero ¿por qué estamos viendo las líneas en blanco adicionales? Esto se debe a ese carácter invisible de *nueva línea*. Cada una de las líneas termina con un salto de línea, así que la sentencia `print` imprime la cadena de la variable *line*, que incluye un salto de línea, y luego `print` añade *otro* salto de línea, resultando en el efecto de doble espacio que vemos.

Podríamos usar el rebanado de líneas para imprimir todos los caracteres menos el último, pero un enfoque más simple es usar el método *rstrip*, que elimina los espacios en blanco del lado derecho de una cadena, de la siguiente manera (`search2.py`):


````python
fhand = open('mbox-short.txt')
for line in fhand:
    line = line.rstrip()
    if line.startswith('From:'):
        print(line)
````

Cuando este programa se ejecuta, obtenemos la siguiente salida:

```
From: stephen.marquard@uct.ac.za
From: louis@media.berkeley.edu
From: zqian@umich.edu
From: rjlowe@iupui.edu
From: zqian@umich.edu
From: rjlowe@iupui.edu
From: cwen@iupui.edu
...
```

A medida que tus programas de procesamiento de archivos se vuelvan más complicados, quizás quieras estructurar tus bucles de búsqueda usando `continue`. La idea básica del bucle de búsqueda es que estás buscando líneas "interesantes" y efectivamente saltando las líneas "no interesantes". Y entonces, cuando encontramos una línea interesante, hacemos algo con esa línea.

Podemos estructurar el bucle para seguir el patrón de saltar las líneas no interesantes de la siguiente manera (`search3.py`):


````python
fhand = open('mbox-short.txt')
for line in fhand:
    line = line.rstrip()
    # Skip 'uninteresting lines'
    if not line.startswith('From:'):
        continue
    # Process our 'interesting' line
    print(line)
````

La salida del programa es la misma. En lenguaje natural, las líneas no interesantes son las que no comienzan con "From:", que saltamos usando `continue`. Para las líneas "interesantes" (es decir, las que comienzan con "From:") realizamos el procesamiento.

Podemos usar el método de cadenas `find` para simular una búsqueda de editor de texto que encuentra líneas donde la cadena de búsqueda está en cualquier parte de la línea. Dado que `find` busca una ocurrencia de una cadena dentro de otra cadena y devuelve la posición de la cadena o -1 si la cadena no se encontró, podemos escribir el siguiente bucle para mostrar las líneas que contienen la cadena "@uct.ac.za" (es decir, provienen de la Universidad de Ciudad del Cabo en Sudáfrica) (`search4.py`):


````python
fhand = open('mbox-short.txt')
for line in fhand:
    line = line.rstrip()
    if line.find('@uct.ac.za') == -1: continue
    print(line)
````

Lo cual produce la siguiente salida:

```
From stephen.marquard@uct.ac.za Sat Jan  5 09:14:16 2008
X-Authentication-Warning: set sender to stephen.marquard@uct.ac.za using -f
From: stephen.marquard@uct.ac.za
Author: stephen.marquard@uct.ac.za
From david.horwitz@uct.ac.za Fri Jan  4 07:02:32 2008
X-Authentication-Warning: set sender to david.horwitz@uct.ac.za using -f
From: david.horwitz@uct.ac.za
Author: david.horwitz@uct.ac.za
...
```

Aquí también usamos la forma contraída de la sentencia `if`, donde ponemos el `continue` en la misma línea que el `if`. Esta forma contraída del `if` funciona igual que si el `continue` estuviera en la línea siguiente e indentado.


## Dejar que el usuario elija el nombre del archivo

En realidad no queremos tener que editar nuestro código Python cada vez que queremos procesar un archivo diferente. Sería más útil pedir al usuario que introduzca la cadena del nombre del archivo cada vez que el programa se ejecuta, de modo que puedan usar nuestro programa sobre diferentes archivos sin cambiar el código Python.

Esto es bastante simple de hacer leyendo el nombre del archivo del usuario usando `input` de la siguiente manera (`search6.py`):


````python
fname = input('Enter the file name: ')
fhand = open(fname)
count = 0
for line in fhand:
    if line.startswith('Subject:'):
        count = count + 1
print('There were', count, 'subject lines in', fname)
````

Leemos el nombre del archivo del usuario y lo colocamos en una variable llamada `fname`, y abrimos ese archivo. Ahora podemos ejecutar el programa repetidamente sobre diferentes archivos.

```
python search6.py
Enter the file name: mbox.txt
There were 1797 subject lines in mbox.txt

python search6.py
Enter the file name: mbox-short.txt
There were 27 subject lines in mbox-short.txt
```

## Usando `try`, `except` y `open`

Te dije que no miraras. Esta es tu última oportunidad.

¿Qué pasa si nuestro usuario escribe algo que no es un nombre de archivo?

```
python search6.py
Enter the file name: missing.txt

  File "search6.py", line 2, in <module>
    fhand = open(fname)

python search6.py

Traceback (most recent call last):
  File "search6.py", line 2, in <module>
    fhand = open(fname)
FileNotFoundError: [Errno 2] No such file or directory: 'na na boo boo'
```

No te rías. Los usuarios eventualmente harán todo lo posible que puedan hacer para romper tus programas, ya sea por error o con intención maliciosa. De hecho, una parte importante de cualquier equipo de desarrollo de software es una persona o grupo llamado *Aseguramiento de la Calidad* (Quality Assurance, o QA para abreviar), cuyo trabajo consiste en hacer las cosas más disparatadas posibles en un intento de romper el software que el programador ha creado.

El equipo de QA es responsable de encontrar los defectos de los programas antes de que hayamos entregado el programa a los usuarios finales, que pueden estar comprando el software o pagando nuestro salario para escribir el software. Así que el equipo de QA es el mejor amigo del programador.

Así que ahora que vemos el defecto en el programa, podemos arreglarlo elegantemente usando la estructura `try`/`except`. Necesitamos asumir que la llamada a `open` podría fallar y añadir código de recuperación para cuando el `open` falla, de la siguiente manera (`search7.py`):


````python
fname = input('Enter the file name: ')
try:
    fhand = open(fname)
except:
    print('File cannot be opened:', fname)
    exit()
count = 0
for line in fhand:
    if line.startswith('Subject:'):
        count = count + 1
print('There were', count, 'subject lines in', fname)
````

La función `exit` termina el programa. Es una función que llamamos que nunca retorna. Ahora, cuando nuestro usuario (o el equipo de QA) escribe tonterías o nombres de archivo incorrectos, los "capturamos" y nos recuperamos con elegancia:

```
python search7.py
Enter the file name: mbox.txt
There were 1797 subject lines in mbox.txt

python search7.py
Enter the file name: na na boo boo
File cannot be opened: na na boo boo
```

Proteger la llamada a `open` es un buen ejemplo del uso adecuado de `try` y `except` en un programa de Python. Usamos el término "Pythonic" (pitónico) cuando estamos haciendo algo a la "manera Python". Podríamos decir que el ejemplo anterior es la forma pitónica de abrir un archivo.

Una vez que tengas más destreza en Python, puedes participar en debates con otros programadores de Python para decidir cuál de dos soluciones equivalentes a un problema es "más pitónica". El objetivo de ser "más pitónico" captura la noción de que la programación es parte ingeniería y parte arte. ## Escribir archivos

Para escribir un archivo, tienes que abrirlo con el modo "w" como segundo parámetro:


````python
fout = open('output.txt', 'w')
print(fout)   # <_io.TextIOWrapper name='output.txt' mode='w' encoding='cp1252'>
````

¡Si el archivo ya existe, abrirlo en modo de escritura borra los datos antiguos y empieza de cero, así que ten cuidado! Si el archivo no existe, se crea uno nuevo.

El método `write` del objeto manejador de archivo pone datos en el archivo, devolviendo el número de caracteres escritos. El modo de escritura por defecto es texto, para escribir (y leer) cadenas.


````python
line1 = "This here's the wattle,\n"
fout.write(line1)   # 24
````

De nuevo, el objeto de archivo lleva la cuenta de dónde está, así que si llamas a `write` de nuevo, añade los nuevos datos al final.

Debemos asegurarnos de gestionar los finales de línea mientras escribimos en el archivo, insertando explícitamente el carácter de nueva línea cuando queremos terminar una línea. La sentencia `print` añade automáticamente un salto de línea, pero el método `write` no añade el salto de línea automáticamente.


````python
line2 = 'the emblem of our land.\n'
fout.write(line2)   # 24
````

Cuando terminas de escribir, tienes que cerrar el archivo para asegurarte de que el último bit de datos se escribe físicamente en el disco, de modo que no se pierda si se va la energía.


````python
fout.close()
````

También podríamos cerrar los archivos que abrimos para lectura, pero podemos ser un poco descuidados si solo estamos abriendo unos pocos archivos, ya que Python se asegura de que todos los archivos abiertos se cierren cuando el programa termina. Cuando estamos escribiendo archivos, queremos cerrar explícitamente los archivos para no dejar nada al azar.


## Depuración

Cuando estás leyendo y escribiendo archivos, podrías tener problemas con los espacios en blanco. Estos errores pueden ser difíciles de depurar porque los espacios, los tabuladores y los saltos de línea son normalmente invisibles:


````python
s = '1 2\t 3\n 4'
print(s)
````

La función incorporada `repr` puede ayudar. Toma cualquier objeto como argumento y devuelve una representación en cadena del objeto. Para las cadenas, representa los caracteres de espacio en blanco con secuencias de barra invertida:


````python
print(repr(s))   # '1 2\t 3\n 4'
````

Esto puede ser útil para la depuración.

Otro problema con el que podrías encontrarte es que diferentes sistemas usan diferentes caracteres para indicar el final de una línea. Algunos sistemas usan un salto de línea, representado `\n`. Otros usan un carácter de retorno, representado `\r`. Algunos usan ambos. Si mueves archivos entre diferentes sistemas, estas inconsistencias podrían causar problemas.

Para la mayoría de los sistemas, hay aplicaciones para convertir de un formato a otro. Puedes encontrarlas (y leer más sobre este tema) en [https://www.wikipedia.org/wiki/Newline](https://wikipedia.org/wiki/Newline). O, por supuesto, podrías escribir una tú mismo.


## Glosario

- **capturar (catch)**: Evitar que una excepción termine un programa usando las sentencias `try` y `except`.

- **nueva línea (newline)**: Un carácter especial usado en archivos y cadenas para indicar el final de una línea.

- **Pythonic (pitónico)**: Una técnica que funciona elegantemente en Python. "Usar try y except es la forma *pitónica* de recuperarse de archivos inexistentes".

- **Aseguramiento de la Calidad (Quality Assurance)**: Una persona o equipo centrado en asegurar la calidad general de un producto de software. QA suele estar involucrado en probar un producto e identificar problemas antes de que el producto se publique.

- **archivo de texto (text file)**: Una secuencia de caracteres almacenada en un almacenamiento permanente como un disco duro.
