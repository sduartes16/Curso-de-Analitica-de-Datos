## Una lista es una secuencia

Como una cadena, una *lista* es una secuencia de valores. En una cadena, los valores son caracteres; en una lista, pueden ser de cualquier tipo. Los valores de las listas se llaman *elementos* o a veces *ítems*.

Hay varias formas de crear una nueva lista; la más simple es encerrar los elementos entre corchetes ("[" y "]"):


````python
[10, 20, 30, 40]
['crunchy frog', 'ram bladder', 'lark vomit']
````

El primer ejemplo es una lista de cuatro enteros. El segundo es una lista de tres cadenas. Los elementos de una lista no tienen que ser del mismo tipo. La siguiente lista contiene una cadena, un float, un entero y (¡mira!) otra lista:


````python
['spam', 2.0, 5, [10, 20]]
````

Una lista dentro de otra lista está *anidada*.

Una lista que no contiene elementos se llama lista vacía; puedes crear una con corchetes vacíos, `[]`.

Como era de esperar, puedes asignar valores de lista a variables:


````python
cheeses = ['Cheddar', 'Edam', 'Gouda']
numbers = [17, 123]
empty = []
print(cheeses, numbers, empty)   # ['Cheddar', 'Edam', 'Gouda'] [17, 123] []
````

## Las listas son mutables

La sintaxis para acceder a los elementos de una lista es la misma que para acceder a los caracteres de una cadena: el operador de corchetes. La expresión dentro de los corchetes especifica el índice. Recuerda que los índices empiezan en 0:


````python
print(cheeses[0])   # Cheddar
````

A diferencia de las cadenas, las listas son mutables porque puedes cambiar el orden de los ítems de una lista o reasignar un ítem de una lista. Cuando el operador de corchetes aparece en el lado izquierdo de una asignación, identifica el elemento de la lista que será asignado.


````python
numbers = [17, 123]
numbers[1] = 5
print(numbers)   # [17, 5]
````

El elemento uno-ésimo de `numbers`, que solía ser 123, ahora es 5.

Puedes pensar en una lista como una relación entre índices y elementos. Esta relación se llama un *mapeo* (mapping); cada índice "mapea a" uno de los elementos.

Los índices de las listas funcionan de la misma manera que los índices de las cadenas:

- Cualquier expresión entera puede usarse como índice.

- Si intentas leer o escribir un elemento que no existe, obtienes un `IndexError`.

- Si un índice tiene un valor negativo, cuenta hacia atrás desde el final de la lista.

El operador `in` también funciona con listas.


````python
cheeses = ['Cheddar', 'Edam', 'Gouda']
'Edam' in cheeses   # True
'Brie' in cheeses   # False
````

## Recorrer una lista

La forma más común de recorrer los elementos de una lista es con un bucle `for`. La sintaxis es la misma que para las cadenas:


````python
for cheese in cheeses:
    print(cheese)
````

Esto funciona bien si solo necesitas leer los elementos de la lista. Pero si quieres escribir o actualizar los elementos, necesitas los índices. Una forma común de hacerlo es combinar las funciones `range` y `len`:


````python
numbers = [17, 123]

for i in range(len(numbers)):
    numbers[i] = numbers[i] * 2

print(numbers)   # [34, 246]
````

Este bucle recorre la lista y actualiza cada elemento. `len` devuelve el número de elementos de la lista. `range` devuelve una lista de índices de 0 a $n-1$, donde $n$ es la longitud de la lista. Cada vez que pasa por el bucle, `i` obtiene el índice del siguiente elemento. La sentencia de asignación en el cuerpo usa `i` para leer el valor anterior del elemento y asignar el nuevo valor.

Un bucle `for` sobre una lista vacía nunca ejecuta el cuerpo:


````python
for x in empty:
    print('This never happens.')

print('(el bucle anterior nunca imprimió nada porque la lista está vacía)')
````

Aunque una lista puede contener otra lista, la lista anidada sigue contando como un solo elemento. La longitud de esta lista es cuatro:


````python
len(['spam', 1, ['Brie', 'Roquefort', 'Pol le Veq'], [1, 2, 3]])   # 4
````

## Operaciones con listas

El operador `+` concatena listas:


````python
a = [1, 2, 3]
b = [4, 5, 6]
c = a + b
print(c)   # [1, 2, 3, 4, 5, 6]
````

De forma similar, el operador `*` repite una lista un número dado de veces:


````python
[0] * 4          # [0, 0, 0, 0]
[1, 2, 3] * 3    # [1, 2, 3, 1, 2, 3, 1, 2, 3]
````

El primer ejemplo repite cuatro veces. El segundo ejemplo repite la lista tres veces.


## Rebanadas de listas

El operador de rebanado también funciona con listas:


````python
t = ['a', 'b', 'c', 'd', 'e', 'f']
t[1:3]   # ['b', 'c']
t[:4]    # ['a', 'b', 'c', 'd']
t[3:]    # ['d', 'e', 'f']
````

Si omites el primer índice, la rebanada empieza al principio. Si omites el segundo, la rebanada llega hasta el final. Así que si omites ambos, la rebanada es una copia de la lista completa.


````python
t[:]   # ['a', 'b', 'c', 'd', 'e', 'f']
````

Dado que las listas son mutables, a menudo es útil hacer una copia antes de realizar operaciones que transforman o mutilan listas.

Un operador de rebanado en el lado izquierdo de una asignación puede actualizar múltiples elementos:


````python
t = ['a', 'b', 'c', 'd', 'e', 'f']
t[1:3] = ['x', 'y']
print(t)   # ['a', 'x', 'y', 'd', 'e', 'f']
````

## Métodos de las listas

Python proporciona métodos que operan sobre listas. Por ejemplo, `append` añade un nuevo elemento al final de una lista:


````python
t = ['a', 'b', 'c']
t.append('d')
print(t)   # ['a', 'b', 'c', 'd']
````

`extend` toma una lista como argumento y añade todos sus elementos:


````python
t1 = ['a', 'b', 'c']
t2 = ['d', 'e']
t1.extend(t2)
print(t1)   # ['a', 'b', 'c', 'd', 'e']
````

Este ejemplo deja `t2` sin modificar.

`sort` ordena los elementos de la lista de menor a mayor:


````python
t = ['d', 'c', 'e', 'b', 'a']
t.sort()
print(t)   # ['a', 'b', 'c', 'd', 'e']
````

La mayoría de los métodos de listas son vacíos (void); modifican la lista y devuelven `None`. Si accidentalmente escribes `t = t.sort()`, te llevarás una decepción con el resultado.


## Eliminar elementos

Hay varias formas de eliminar elementos de una lista. Si conoces el índice del elemento que quieres, puedes usar `pop`:


````python
t = ['a', 'b', 'c']
x = t.pop(1)
print(t)   # ['a', 'c']
print(x)   # b
````

`pop` modifica la lista y devuelve el elemento que fue eliminado. Si no proporcionas un índice, elimina y devuelve el último elemento.

Si no necesitas el valor eliminado, puedes usar la sentencia `del`:


````python
t = ['a', 'b', 'c']
del t[1]
print(t)   # ['a', 'c']
````

Si conoces el elemento que quieres eliminar (pero no el índice), puedes usar `remove`:


````python
t = ['a', 'b', 'c']
t.remove('b')
print(t)   # ['a', 'c']
````

El valor de retorno de `remove` es `None`.

Para eliminar más de un elemento, puedes usar `del` con un índice de rebanada:


````python
t = ['a', 'b', 'c', 'd', 'e', 'f']
del t[1:5]
print(t)   # ['a', 'f']
````

Como de costumbre, la rebanada selecciona todos los elementos hasta el segundo índice, pero sin incluirlo.


## Listas y funciones

Hay un número de funciones incorporadas que pueden usarse sobre listas y que te permiten examinar rápidamente una lista sin escribir tus propios bucles:


````python
nums = [3, 41, 12, 9, 74, 15]
print(len(nums))              # 6
print(max(nums))              # 74
print(min(nums))              # 3
print(sum(nums))              # 154
print(sum(nums)/len(nums))    # 25
````

La función `sum()` solo funciona cuando los elementos de la lista son números. Las demás funciones (`max()`, `len()`, etc.) funcionan con listas de cadenas y otros tipos que se puedan comparar.

Podríamos reescribir un programa anterior que calculaba el promedio de una lista de números introducidos por el usuario usando una lista.

Primero, el programa para calcular un promedio sin una lista (`avenum.py`):


````python
total = 0
count = 0
while (True):
    inp = input('Enter a number: ')
    if inp == 'done': break
    value = float(inp)
    total = total + value
    count = count + 1

average = total / count
print('Average:', average)
````

En este programa, tenemos las variables `count` y `total` para llevar la cantidad y el total acumulado de los números del usuario mientras le pedimos repetidamente un número.

Podríamos simplemente recordar cada número a medida que el usuario lo introduce y usar funciones incorporadas para calcular la suma y la cuenta al final (`avelist.py`):


````python
numlist = list()
while (True):
    inp = input('Enter a number: ')
    if inp == 'done': break
    value = float(inp)
    numlist.append(value)

average = sum(numlist) / len(numlist)
print('Average:', average)
````

Creamos una lista vacía antes de que comience el bucle, y luego cada vez que tenemos un número, lo añadimos a la lista. Al final del programa, simplemente calculamos la suma de los números de la lista y la dividimos entre la cuenta de los números de la lista para obtener el promedio.


## Listas y cadenas

Una cadena es una secuencia de caracteres y una lista es una secuencia de valores, pero una lista de caracteres no es lo mismo que una cadena. Para convertir de una cadena a una lista de caracteres, puedes usar `list`:


````python
s = 'spam'
t = list(s)
print(t)   # ['s', 'p', 'a', 'm']
````

Como `list` es el nombre de una función incorporada, deberías evitar usarlo como nombre de variable. Yo también evito la letra "l" porque se parece demasiado al número "1". Por eso uso "t".

La función `list` rompe una cadena en letras individuales. Si quieres romper una cadena en palabras, puedes usar el método `split`:


````python
s = 'pining for the fjords'
t = s.split()
print(t)      # ['pining', 'for', 'the', 'fjords']
print(t[2])   # the
````

Una vez que has usado `split` para romper la cadena en una lista de palabras, puedes usar el operador de índice (corchete) para mirar una palabra particular de la lista.

Puedes llamar a `split` con un argumento opcional llamado *delimitador* que especifica qué caracteres usar como límites de palabras. El siguiente ejemplo usa un guion como delimitador:


````python
s = 'spam-spam-spam'
delimiter = '-'
s.split(delimiter)   # ['spam', 'spam', 'spam']
````

`join` es el inverso de `split`. Toma una lista de cadenas y concatena los elementos. `join` es un método de cadenas, así que tienes que invocarlo sobre el delimitador y pasar la lista como parámetro:


````python
t = ['pining', 'for', 'the', 'fjords']
delimiter = ' '
delimiter.join(t)   # 'pining for the fjords'
````

En este caso el delimitador es un carácter de espacio, así que `join` pone un espacio entre las palabras. Para concatenar cadenas sin espacios, puedes usar la cadena vacía, "", como delimitador.

## Objetos y valores

Si ejecutamos estas sentencias de asignación:


````python
a = 'banana'
b = 'banana' 
````

sabemos que `a` y `b` se refieren a una cadena, pero no sabemos si se refieren a la *misma* cadena. Hay dos estados posibles:


![Variables y objetos](../../images/list1.svg)


En un caso, `a` y `b` se refieren a dos objetos diferentes que tienen el mismo valor. En el segundo caso, se refieren al mismo objeto.

Para comprobar si dos variables se refieren al mismo objeto, puedes usar el operador `is`.


````python
a = 'banana'
b = 'banana'
a is b   # True
````

En este ejemplo, Python solo creó un objeto de cadena, y tanto `a` como `b` se refieren a él.

Pero cuando creas dos listas, obtienes dos objetos:


````python
a = [1, 2, 3]
b = [1, 2, 3]
a is b   # False

````

En este caso diríamos que las dos listas son *equivalentes*, porque tienen los mismos elementos, pero no *idénticas*, porque no son el mismo objeto. Si dos objetos son idénticos, también son equivalentes, pero si son equivalentes, no son necesariamente idénticos.

Hasta ahora, hemos estado usando "objeto" y "valor" indistintamente, pero es más preciso decir que un objeto tiene un valor. Si ejecutas `a = [1,2,3]`, `a` se refiere a un objeto lista cuyo valor es una secuencia particular de elementos. Si otra lista tiene los mismos elementos, diríamos que tiene el mismo valor.


## Alias (aliasing)

Si `a` se refiere a un objeto y asignas `b = a`, entonces ambas variables se refieren al mismo objeto:


````python
a = [1, 2, 3]
b = a
b is a   # True
````

La asociación de una variable con un objeto se llama una *referencia*. En este ejemplo, hay dos referencias al mismo objeto.

Un objeto con más de una referencia tiene más de un nombre, así que decimos que el objeto tiene *alias*.

Si el objeto con alias es mutable, los cambios hechos con un alias afectan al otro:


````python
b[0] = 17
print(a)   # [17, 2, 3]
````

Aunque este comportamiento puede ser útil, es propenso a errores. En general, es más seguro evitar los alias cuando trabajas con objetos mutables.

Para objetos inmutables como las cadenas, los alias no suponen tanto problema. En este ejemplo:

```
a = 'banana'
b = 'banana'
```

casi nunca hace diferencia si `a` y `b` se refieren a la misma cadena o no.


## Listas como argumentos

Cuando pasas una lista a una función, la función obtiene una referencia a la lista. Si la función modifica un parámetro de lista, quien la llamó ve el cambio. Por ejemplo, `delete_head` elimina el primer elemento de una lista:


````python
def delete_head(t):
    del t[0]
````

Así es como se usa:


````python
letters = ['a', 'b', 'c']
delete_head(letters)
print(letters)   # ['b', 'c']
````

El parámetro `t` y la variable `letters` son alias del mismo objeto.

Es importante distinguir entre operaciones que modifican listas y operaciones que crean listas nuevas. Por ejemplo, el método `append` modifica una lista, pero el operador `+` crea una lista nueva:


````python
t1 = [1, 2]
t2 = t1.append(3)
print(t1)   # [1, 2, 3]
print(t2)   # None
````

````python
t3 = t1 + [3]
print(t3)   # [1, 2, 3]
t1 is t3    # False
````

Esta diferencia es importante cuando escribes funciones que se supone que modifican listas. Por ejemplo, esta función *no* elimina la cabeza de una lista:

```python
def bad_delete_head(t):
    t = t[1:]              # ¡INCORRECTO!
```

El operador de rebanado crea una nueva lista y la asignación hace que `t` se refiera a ella, pero nada de eso tiene ningún efecto sobre la lista que se pasó como argumento.

Una alternativa es escribir una función que crea y devuelve una nueva lista. Por ejemplo, `tail` devuelve todos los elementos menos el primero de una lista:


````python
def tail(t):
    return t[1:]
````

Esta función deja la lista original sin modificar. Así es como se usa:


````python
letters = ['a', 'b', 'c']
rest = tail(letters)
print(rest)   # ['b', 'c']
````

## Depuración

El uso descuidado de las listas (y otros objetos mutables) puede llevar a largas horas de depuración. Aquí hay algunas trampas comunes y formas de evitarlas:

**1. No olvides que la mayoría de los métodos de listas modifican el argumento y devuelven `None`.** Esto es lo contrario de los métodos de cadenas, que devuelven una nueva cadena y dejan la original intacta.

Si estás acostumbrado a escribir código de cadenas como este:

```
word = word.strip()
```

Es tentador escribir código de listas como este:

```
t = t.sort()           # ¡INCORRECTO!
```

Como `sort` devuelve `None`, es probable que la siguiente operación que realices con `t` falle.

Antes de usar métodos y operadores de listas, deberías leer la documentación cuidadosamente y luego probarlos en modo interactivo. Los métodos y operadores que las listas comparten con otras secuencias (como las cadenas) están documentados en:

[docs.python.org/library/stdtypes.html#common-sequence-operations](https://docs.python.org/library/stdtypes.html#common-sequence-operations)

Los métodos y operadores que solo se aplican a secuencias mutables están documentados en:

[docs.python.org/library/stdtypes.html#mutable-sequence-types](https://docs.python.org/library/stdtypes.html#mutable-sequence-types)

**2. Elige un idioma (idiom) y apégate a él.**

Parte del problema con las listas es que hay demasiadas formas de hacer las cosas. Por ejemplo, para eliminar un elemento de una lista, puedes usar `pop`, `remove`, `del`, o incluso una asignación de rebanada.

Para añadir un elemento, puedes usar el método `append` o el operador `+`. Pero no olvides que estas son correctas:

```
t.append(x)
t = t + [x]
```

Y estas son incorrectas:

```
t.append([x])          # ¡INCORRECTO!
t = t.append(x)        # ¡INCORRECTO!
t + [x]                # ¡INCORRECTO!
t = t + x              # ¡INCORRECTO!
```

Prueba cada uno de estos ejemplos en modo interactivo para asegurarte de que entiendes lo que hacen. Observa que solo el último causa un error en tiempo de ejecución; los otros tres son legales, pero hacen lo incorrecto.

**3. Haz copias para evitar los alias.**

Si quieres usar un método como `sort` que modifica el argumento, pero necesitas mantener también la lista original, puedes hacer una copia.

```
orig = t[:]
t.sort()
```

En este ejemplo también podrías usar la función incorporada `sorted`, que devuelve una nueva lista ordenada y deja la original intacta. ¡Pero en ese caso deberías evitar usar `sorted` como nombre de variable!

**4. Listas, `split` y archivos**

Cuando leemos y analizamos archivos, hay muchas oportunidades de encontrar entradas que pueden hacer fallar nuestro programa, así que es una buena idea revisitar el patrón *guardián* cuando se trata de escribir programas que leen un archivo y buscan una "aguja en un pajar".

Revisemos nuestro programa que busca el día de la semana en las líneas "From" de nuestro archivo:

```
From stephen.marquard@uct.ac.za Sat Jan  5 09:14:16 2008
```

Dado que estamos dividiendo esta línea en palabras, podríamos prescindir del uso de `startswith` y simplemente mirar la primera palabra de la línea para determinar si nos interesa la línea en absoluto. Podemos usar `continue` para saltar las líneas que no tienen "From" como primera palabra, de la siguiente manera:

```
fhand = open('mbox-short.txt')
for line in fhand:
    words = line.split()
    if words[0] != 'From' : continue
    print(words[2])
```

Esto se ve mucho más simple y ni siquiera necesitamos hacer el `rstrip` para eliminar el salto de línea al final del archivo. ¿Pero es mejor?

```
python search8.py
Sat
Traceback (most recent call last):
  File "search8.py", line 5, in <module>
    if words[0] != 'From' : continue
IndexError: list index out of range
```

Funciona parcialmente y vemos el día de la primera línea (Sat), pero luego el programa falla con un error de traceback. ¿Qué salió mal? ¿Qué datos desordenados hicieron fallar a nuestro elegante, ingenioso y muy pitónico programa?

Podrías quedarte mirándolo durante mucho tiempo y tratar de descifrarlo o pedirle ayuda a alguien, pero el enfoque más rápido e inteligente es añadir una sentencia `print`. El mejor lugar para añadir la sentencia print es justo antes de la línea donde el programa falló, e imprimir los datos que parecen estar causando el fallo.

Ahora bien, este enfoque puede generar muchas líneas de salida, pero al menos tendrás inmediatamente alguna pista sobre el problema en cuestión. Así que añadimos un print de la variable `words` justo antes de la línea cinco. Incluso añadimos un prefijo "Debug:" a la línea para poder mantener nuestra salida regular separada de nuestra salida de depuración.

```
for line in fhand:
    words = line.split()
    print('Debug:', words)
    if words[0] != 'From' : continue
    print(words[2])
```

Cuando ejecutamos el programa, mucha salida se desplaza por la pantalla, pero al final vemos nuestra salida de depuración y el traceback, así que sabemos qué ocurrió justo antes del traceback.

```
Debug: ['X-DSPAM-Confidence:', '0.8475']
Debug: ['X-DSPAM-Probability:', '0.0000']
Debug: []
Traceback (most recent call last):
  File "search9.py", line 6, in <module>
    if words[0] != 'From' : continue
IndexError: list index out of range
```

Cada línea de depuración está imprimiendo la lista de palabras que obtenemos cuando hacemos `split` de la línea en palabras. Cuando el programa falla, la lista de palabras está vacía `[]`. Si abrimos el archivo en un editor de texto y miramos el archivo, en ese punto se ve de la siguiente manera:

```
X-DSPAM-Result: Innocent
X-DSPAM-Processed: Sat Jan  5 09:14:16 2008
X-DSPAM-Confidence: 0.8475
X-DSPAM-Probability: 0.0000

Details: http://source.sakaiproject.org/viewsvn/?view=rev&rev=39772
```

¡El error ocurre cuando nuestro programa encuentra una línea en blanco! Por supuesto, hay "cero palabras" en una línea en blanco. ¿Por qué no pensamos en eso cuando estábamos escribiendo el código? Cuando el código busca la primera palabra (`word[0]`) para comprobar si coincide con "From", obtenemos un error de "index out of range".

Este es, por supuesto, el lugar perfecto para añadir algo de código *guardián* para evitar comprobar la primera palabra si la primera palabra no está ahí. Hay muchas formas de proteger este código; elegiremos comprobar el número de palabras que tenemos antes de mirar la primera palabra:

## Glosario

- **alias (aliasing)**: Una circunstancia en la que dos o más variables se refieren al mismo objeto.

- **delimitador (delimiter)**: Un carácter o cadena usado para indicar dónde debe dividirse una cadena.

- **elemento (element)**: Uno de los valores de una lista (u otra secuencia); también llamados ítems.

- **equivalente (equivalent)**: Que tiene el mismo valor.

- **índice (index)**: Un valor entero que indica un elemento en una lista.

- **idéntico (identical)**: Ser el mismo objeto (lo que implica equivalencia).

- **lista (list)**: Una secuencia de valores.

- **recorrido de lista (list traversal)**: El acceso secuencial a cada elemento de una lista.

- **lista anidada (nested list)**: Una lista que es elemento de otra lista.

- **objeto (object)**: Algo a lo que una variable puede referirse. Un objeto tiene un tipo y un valor.

- **referencia (reference)**: La asociación entre una variable y su valor.
