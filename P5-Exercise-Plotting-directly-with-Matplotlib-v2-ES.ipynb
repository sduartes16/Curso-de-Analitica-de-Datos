# Gráfico de líneas<a id="6"></a>


Un gráfico de líneas muestra la relación entre dos variables continuas a lo largo de un intervalo continuo, mostrando la tendencia o el patrón de los datos.<br><br>Creemos un gráfico de líneas para visualizar la tendencia de los inmigrantes (a Canadá) durante 1980 a 2013.<br>Necesitamos el total de inmigrantes por año.<br><br>Crearemos un nuevo dataframe solo con las columnas que contienen los años<br> y luego aplicaremos sum() al dataframe<br><br>Puedes crear un gráfico de líneas directamente sobre los **ejes** llamando a la función **plot()**


````python
#Como years está en formato de arreglo (array), tendrás que mapearlo a str para graficar
#y=list(map(str, years))

#creando el df solo con las columnas de años desde 1980 - 2013
df_line=df_can[years]

#Aplicando sum para obtener el total de inmigrantes por año
total_immigrants=df_line.sum()
total_immigrants
````

````python
#Crear figura y ejes
fig, ax = plt.subplots()

# Graficar la línea
ax.plot(total_immigrants)

#Configurar el título
ax.set_title('Immigrants between 1980 to 2013') 
#Configurar las etiquetas
ax.set_xlabel('Years')
ax.set_ylabel('Total Immigrants')

#Mostrar el gráfico
plt.show()
````

#### La función plot llenó el eje x con los valores del índice (los años) y el eje y con los valores de la columna (la población). <br>Sin embargo, observa que los años no se mostraban porque son de tipo texto (string).<br> <br>Por lo tanto, cambiemos el tipo de los valores del índice a entero para graficar.


````python
#Crear figura y ejes
fig, ax = plt.subplots()

#Cambiar el tipo del índice a entero
total_immigrants.index = total_immigrants.index.map(int)

# Graficar la línea
ax.plot(total_immigrants)

#Configurar el título
ax.set_title('Immigrants between 1980 to 2013') 

#Configurar las etiquetas
ax.set_xlabel('Years')
ax.set_ylabel('Total Immigrants')

#Mostrar el gráfico
plt.show()
````

# Gráfico de dispersión<a id="8"></a>


Un gráfico de dispersión visualiza la relación entre dos variables continuas, mostrando puntos de datos individuales como puntos en un plano bidimensional, lo que permite examinar patrones, agrupaciones (clústeres) y correlaciones.<br><br>Creemos un *gráfico de dispersión* para visualizar la tendencia de los inmigrantes (a Canadá) durante 1980 a 2013.<br>Necesitamos el total de inmigrantes por año.<br><br>Crearemos un nuevo dataframe solo con las columnas que contienen los años<br> y luego aplicaremos sum() al dataframe<br><br>Puedes crear un gráfico de dispersión directamente sobre **ax** llamando a la función **scatter()**


````python
#Crear figura y ejes
fig, ax = plt.subplots(figsize=(8, 4))

# Personalizar el gráfico de dispersión 
ax.scatter(years, total_immigrants, 
           marker='o', #configurar los marcadores
           s = 20, #configurar el tamaño de los marcadores
           color='darkblue')#el color del marcador

#añadir el título 
plt.title('Immigrants between 1980 to 2013') 
#añadir las etiquetas 
plt.xlabel('Years')
plt.ylabel('Total Immigrants') 
#incluir la cuadrícula
plt.grid(True)

#Leyenda en la parte superior central de la figura
ax.legend(["Immigrants"], loc='upper center')

#Mostrar el gráfico
plt.show()
````

# Gráfico de barras<a id="10"></a>


Un gráfico de barras representa datos categóricos con barras rectangulares, donde la altura de cada barra corresponde al valor de una categoría específica, lo que lo hace adecuado para comparar valores entre distintas categorías.<br><br>Creemos un gráfico de barras para visualizar los 5 países que más contribuyeron a la inmigración a Canadá entre 1980 y 2013.
<br><br>Aplica la función sort_values sobre la columna 'Total' de nuestros datos<br><br>Crearemos un nuevo dataframe solo con las columnas que contienen los años<br> luego aplicaremos sum() al dataframe y podremos crear un dataframe aparte para los cinco países principales<br><br>Además puedes usar los nombres de los países para etiquetar cada barra en el gráfico


````python
#Ordenar el dataframe por 'Total' en orden descendente
df_can.sort_values(['Total'], ascending=False, axis=0, inplace=True)

# obtener las 5 primeras entradas con la función head
df_top5 = df_can.head()

#restableciendo el índice a su forma original
df_bar_5=df_top5.reset_index()

#Creando una lista con los nombres de los 5 países principales
label=list(df_bar_5.Country)
label
````

##### El tercer nombre es demasiado largo para caber en el eje x como etiqueta. Arreglémoslo usando indexación


````python
label[2]='UK'
label
````

````python
fig, ax = plt.subplots(figsize=(10, 4))

ax.bar(label,df_bar_5['Total'], label=label)
ax.set_title('Immigration Trend of Top 5 Countries')
ax.set_ylabel('Number of Immigrants')
ax.set_xlabel('Years')

plt.show()
````

# Histograma<a id="12"></a>


Un histograma es una forma de representar la distribución de *frecuencias* de un conjunto de datos numéricos. Funciona dividiendo el eje x en *bins* (intervalos), asignando cada punto de datos de nuestro conjunto a un intervalo y luego contando cuántos puntos de datos han sido asignados a cada intervalo. Así, el eje y es la frecuencia o el número de puntos de datos en cada intervalo. Ten en cuenta que podemos cambiar el tamaño de los intervalos y normalmente hay que ajustarlo para que la distribución se muestre de forma clara.
<br><br>Averigüemos cuál es la distribución de frecuencias del número (población) de nuevos inmigrantes de los distintos países a Canadá en 2013.<br>


````python
df_country = df_can.groupby(['Country'])['2013'].sum().reset_index()

#Crear figura y ejes
fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(df_country['2013'])
ax.set_title('New Immigrants in 2013') 
ax.set_xlabel('Number of Immigrants')
ax.set_ylabel('Number of Countries')
ax.legend(['Immigrants'])

#Mostrar el gráfico
plt.show()
````

Nuestro gráfico no coincide con las barras<br>Por defecto, el método `histrogram` divide el conjunto de datos en 10 intervalos. La figura siguiente resume los rangos de los intervalos y la distribución de frecuencias de la inmigración en 2013<br> La función hist devuelve una lista de arreglos con 1. los conteos y 2. los intervalos. Podemos obtenerlos usando el desempaquetado (unpacking) <br> y además usar los intervalos como marcas (ticks) del eje x


````python
# Graficar las barras
fig, ax = plt.subplots(figsize=(10, 4))
count = ax.hist(df_country['2013'])

#puedes revisar los arreglos en count con el índice count[0] para los conteos y count[1] para los intervalos

ax.set_title('New Immigrants in 2013') 
ax.set_xlabel('Number of Immigrants')
ax.set_ylabel('Number of Countries')
ax.set_xticks(list(map(int,count[1])))
ax.legend(['Immigrants'])

#Mostrar el gráfico
plt.show()
````

# Gráfico de sectores (pie)<a id="14"></a>

Un gráfico de sectores representa la distribución proporcional o porcentual de distintas categorías de un conjunto de datos mediante sectores de un círculo.
<br> Creemos un gráfico de sectores que represente el 'Total de inmigrantes' para los años 1980 a 1985


````python
fig,ax=plt.subplots()

#Gráfico de sectores sobre los inmigrantes
ax.pie(total_immigrants[0:5], labels=years[0:5], 
       colors = ['gold','blue','lightgreen','coral','cyan'],
       autopct='%1.1f%%',explode = [0,0,0,0,0.1]) #usando explode para resaltar el valor más bajo 

ax.set_aspect('equal')  # Asegurar que el gráfico de sectores se dibuje como un círculo

plt.title('Distribution of Immigrants from 1980 to 1985')
#plt.legend(years[0:5]), incluye la leyenda, si no quieres pasar las etiquetas
plt.show()
````

# Subgráficos<a id="18"></a>


Exploremos cómo mostrar más de un gráfico en la misma figura e indicar el número de filas y columnas a crear en la función subplots.

Por ejemplo, creemos un gráfico de líneas y uno de dispersión en una fila <br>
plt.subplots() <br>
Puedes usar las mismas funciones con las que graficaste los gráficos de líneas y de dispersión al inicio de este laboratorio
<br>
Ambos subgráficos compartirán el mismo eje y, porque los datos del eje y son los mismos. Por eso, asigna el parámetro 'Sharey' como True en el código de abajo. Observa también el uso de 'suptitle'


````python
# Crear una figura con dos ejes en una fila

fig, axs = plt.subplots(1, 2, sharey=True)

#Graficar en el primer eje - el de la izquierda
axs[0].plot(total_immigrants)
axs[0].set_title("Line plot on immigrants")

#Graficar en el segundo eje - el de la derecha
axs[1].scatter(total_immigrants.index, total_immigrants)
axs[1].set_title("Scatter plot on immigrants")

axs[0].set_ylabel("Number of Immigrants")
            
#Añadir un título para la figura completa
fig.suptitle('Subplotting Example', fontsize=15)

# Ajustar el espaciado entre subgráficos
fig.tight_layout()


# Mostrar la figura
plt.show()
````

También puedes implementar los subgráficos con **add_subplot()** como se muestra a continuación:


````python
# Crear una figura con cuatro ejes - dos filas, dos columnas
fig = plt.figure(figsize=(8,4))

# Añadir el primer subgráfico (arriba a la izquierda)
axs1 = fig.add_subplot(1, 2, 1)
#Graficar en el primer eje - el de la izquierda
axs1.plot(total_immigrants)
axs1.set_title("Line plot on immigrants")

# Añadir el segundo subgráfico (arriba a la derecha)
axs2 = fig.add_subplot(1, 2, 2)
#Graficar en el segundo eje - el de la derecha
axs2.barh(total_immigrants.index, total_immigrants) #Observa el uso de 'barh' para crear un gráfico de barras horizontal
axs2.set_title("Bar plot on immigrants")
            
#Añadir un título para la figura completa
fig.suptitle('Subplotting Example', fontsize=15)

# Ajustar el espaciado entre subgráficos
fig.tight_layout()


# Mostrar la figura
plt.show()
````
